## 2. Data Loading และ Data Preparation
* แบ่งข้อมูลแบบ **stratified** 70/15/15 เพื่อให้สัดส่วนแต่ละ class เท่ากันทุกชุด
* validation/test ใช้แค่ resize + normalize เสมอ (augment เฉพาะ train set)

```python
# Section 2: Data Loading and Stratified Split
from pathlib import Path
from collections import Counter
import warnings
import numpy as np
import torch
from torch.utils.data import Dataset
from PIL import Image
from torchvision import transforms
from sklearn.model_selection import train_test_split

ROOT = Path(
    "/kaggle/input/datasets/pkdarabi/"
    "bone-fracture-detection-computer-vision-project/"
    "BoneFractureYolo8"
)
assert ROOT.exists(), f"ไม่พบ Dataset: {ROOT}"

class_names = [
    "elbow positive",
    "fingers positive",
    "forearm fracture",
    "humerus fracture",
    "humerus",
    "shoulder fracture",
    "wrist positive",
]
NUM_CLASSES = len(class_names)

S = CFG["img_size"]
MEAN = [0.485, 0.456, 0.406]
STD = [0.229, 0.224, 0.225]

eval_tf = transforms.Compose([
    transforms.Resize((S, S)),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

# อ่านภาพจาก train, valid, test แล้วรวมไว้เพื่อแบ่งใหม่
extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
all_records = []
all_targets = []
primary_labels = []
source_counts = Counter()

for source_split in ["train", "valid", "test"]:
    image_dir = ROOT / source_split / "images"
    label_dir = ROOT / source_split / "labels"

    if not image_dir.exists():
        raise FileNotFoundError(f"ไม่พบโฟลเดอร์: {image_dir}")

    image_paths = sorted(
        p for p in image_dir.iterdir()
        if p.is_file() and p.suffix.lower() in extensions
    )

    for image_path in image_paths:
        label_path = label_dir / f"{image_path.stem}.txt"
        target = np.zeros(NUM_CLASSES, dtype=np.float32)
        found_ids = []

        if label_path.exists():
            for line in label_path.read_text().splitlines():
                parts = line.split()
                if not parts:
                    continue

                try:
                    class_id = int(parts[0])
                except (ValueError, IndexError):
                    continue

                if 0 <= class_id < NUM_CLASSES:
                    target[class_id] = 1.0
                    if class_id not in found_ids:
                        found_ids.append(class_id)

        all_records.append((image_path, label_path))
        all_targets.append(target)

        # Stratify ด้วยคลาสแรกที่พบใน label
        primary_labels.append(
            f"class_{found_ids[0]}" if found_ids else "no_label"
        )
        source_counts[source_split] += 1

if len(all_records) == 0:
    raise RuntimeError(f"ไม่พบภาพใน {ROOT}")

all_targets = np.asarray(all_targets, dtype=np.float32)
all_indices = np.arange(len(all_records))

def get_safe_strata(indices, test_size):
    labels = [primary_labels[int(i)] for i in indices]
    counts = Counter(labels)

    # รวมกลุ่มที่มีน้อยกว่า 2 ตัวอย่าง
    mapped = [
        label if counts[label] >= 2 else "__rare__"
        for label in labels
    ]
    mapped_counts = Counter(mapped)

    n_test = int(np.ceil(len(indices) * test_size))
    n_train = len(indices) - n_test

    if (
        len(mapped_counts) < 2
        or min(mapped_counts.values()) < 2
        or n_test < len(mapped_counts)
        or n_train < len(mapped_counts)
    ):
        return None

    return mapped

def split_indices(indices, test_size, random_state, name):
    strata = get_safe_strata(indices, test_size)

    if strata is None:
        warnings.warn(
            f"{name}: ใช้ random split เพราะจำนวนบางกลุ่ม "
            "ไม่เพียงพอสำหรับ stratified split"
        )

    left, right = train_test_split(
        indices,
        test_size=test_size,
        random_state=random_state,
        shuffle=True,
        stratify=strata,
    )
    return np.asarray(left), np.asarray(right)

# 70% Train และ 30% ชั่วคราว
train_idx, temp_idx = split_indices(
    all_indices,
    test_size=0.30,
    random_state=SEED,
    name="Train/Temp"
)

# แบ่ง 30% ที่เหลือครึ่งหนึ่งเป็น Validation และ Test
val_idx, test_idx = split_indices(
    temp_idx,
    test_size=0.50,
    random_state=SEED + 1,
    name="Validation/Test"
)

train_records = [all_records[int(i)] for i in train_idx]
val_records = [all_records[int(i)] for i in val_idx]
test_records = [all_records[int(i)] for i in test_idx]

train_targets = all_targets[train_idx]
val_targets = all_targets[val_idx]
test_targets = all_targets[test_idx]

split_indices_map = {
    "Train": train_idx,
    "Validation": val_idx,
    "Test": test_idx,
}

class BoneFractureDataset(Dataset):
    def __init__(self, records, targets, transform=None):
        if len(records) != len(targets):
            raise ValueError("จำนวน records และ targets ไม่ตรงกัน")

        self.records = records
        self.targets = np.asarray(targets, dtype=np.float32)
        self.transform = transform

    def __len__(self):
        return len(self.records)

    def __getitem__(self, index):
        image_path, _ = self.records[index]
        image = Image.open(image_path).convert("RGB")
        target = torch.tensor(
            self.targets[index], dtype=torch.float32
        )

        if self.transform:
            image = self.transform(image)

        return image, target

# ชุดข้อมูลพื้นฐานสำหรับสำรวจภาพ
train_base_ds = BoneFractureDataset(
    train_records, train_targets, eval_tf
)
val_base_ds = BoneFractureDataset(
    val_records, val_targets, eval_tf
)
test_base_ds = BoneFractureDataset(
    test_records, test_targets, eval_tf
)

print("จำนวนภาพต้นทาง:", dict(source_counts))
print("รวมทั้งหมด:", len(all_records))

for name, indices in split_indices_map.items():
    print(
        f"{name}: {len(indices)} ภาพ "
        f"({len(indices) / len(all_records):.1%})"
    )

print("Class names:", class_names)

# ตรวจการกระจายของ label ในแต่ละ split
for split_name, indices in split_indices_map.items():
    counts = all_targets[indices].sum(axis=0).astype(int)
    print(f"\n{split_name}:")
    for class_name, count in zip(class_names, counts):
        print(f"  {class_name}: {count}")
```

## 3. Data Augmentation (กลุ่มออกแบบเอง)

| Augmentation | ค่าที่ใช้ | เหตุผลและข้อควรระวัง |
|---|---|---|
| RandomHorizontalFlip | p = 0.5 | เพิ่มความหลากหลายของภาพ โดยใช้เมื่อการกลับซ้าย-ขวาไม่เปลี่ยนความหมายของคลาส |
| RandomRotation | degrees = 5 | จำลองความเอียงเล็กน้อยจากการจัดวางภาพ โดยไม่หมุนภาพกลับหัว |
| RandomAffine | translate = (0.03, 0.03), scale = (0.95, 1.05) | จำลองการเปลี่ยนตำแหน่งและขนาดเล็กน้อย |

ใช้ Augmentation เฉพาะชุด Train ส่วน Validation และ Test ใช้ Resize และ Normalize เท่านั้น

```python
# Section 3: Transform and DataLoader
from torchvision import transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import numpy as np
import torch

S = CFG["img_size"]
MEAN = [0.485, 0.456, 0.406]
STD = [0.229, 0.224, 0.225]

# Validation/Test ไม่ทำ augmentation
eval_tf = transforms.Compose([
    transforms.Resize((S, S)),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

# ใช้ augmentation เฉพาะ Train ของการทดลอง B
aug_tf = transforms.Compose([
    transforms.Resize((S, S)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=5),
    transforms.RandomAffine(
        degrees=0,
        translate=(0.03, 0.03),
        scale=(0.95, 1.05),
    ),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

# A และ B ใช้ records และ labels ชุดเดียวกัน
train_ds = {
    "A_no_aug": BoneFractureDataset(
        train_records, train_targets, eval_tf
    ),
    "B_aug": BoneFractureDataset(
        train_records, train_targets, aug_tf
    ),
}

val_ds = BoneFractureDataset(
    val_records, val_targets, eval_tf
)
test_ds = BoneFractureDataset(
    test_records, test_targets, eval_tf
)

LOADER_KWARGS = {
    "batch_size": CFG["batch_size"],
    "num_workers": CFG["num_workers"],
    "pin_memory": device.type == "cuda",
}

def make_train_loader(exp):
    if exp not in train_ds:
        raise KeyError(f"ไม่พบ experiment: {exp}")

    return DataLoader(
        train_ds[exp],
        shuffle=True,
        drop_last=False,
        **LOADER_KWARGS,
    )

val_loader = DataLoader(
    val_ds,
    shuffle=False,
    drop_last=False,
    **LOADER_KWARGS,
)

test_loader = DataLoader(
    test_ds,
    shuffle=False,
    drop_last=False,
    **LOADER_KWARGS,
)

print("Train A:", len(train_ds["A_no_aug"]))
print("Train B:", len(train_ds["B_aug"]))
print("Validation:", len(val_ds))
print("Test:", len(test_ds))
print("DataLoaders ready:", device)
```

```python
def denorm(x):
    mean = torch.tensor(MEAN).view(3, 1, 1)
    std = torch.tensor(STD).view(3, 1, 1)

    return (
        (x.detach().cpu() * std + mean)
        .clamp(0, 1)
        .permute(1, 2, 0)
        .numpy()
    )

original_ds = train_ds["A_no_aug"]
augmented_ds = train_ds["B_aug"]

rng = np.random.default_rng(SEED)
sample_ids = rng.choice(
    len(original_ds),
    size=min(3, len(original_ds)),
    replace=False,
)

fig, axes = plt.subplots(3, 4, figsize=(12, 8))

for row, image_id in enumerate(sample_ids):
    original_img, target = original_ds[image_id]

    axes[row, 0].imshow(denorm(original_img))
    axes[row, 0].set_title("Original")

    for col in range(1, 4):
        aug_img, _ = augmented_ds[image_id]
        axes[row, col].imshow(denorm(aug_img))
        axes[row, col].set_title(f"Augmented {col}")

    for ax in axes[row]:
        ax.axis("off")

plt.tight_layout()
plt.show()
```

## 4. Model: Transfer Learning ด้วย ResNet-18
ใช้ pretrained weights จาก ImageNet แล้วเปลี่ยนชั้นสุดท้ายให้เท่ากับจำนวน class
(กลุ่มสามารถเปลี่ยนเป็น EfficientNet, MobileNet, ConvNeXt ฯลฯ ได้ แต่ต้องใช้โมเดลเดียวกันในการทดลอง A และ B)

```python
# Section 4: ResNet-18 from Scratch
import torch
import torch.nn as nn
from torchvision import models

def build_model(pretrained=False):
    # ResNet-18 แบบ Training from Scratch
    # pretrained=False หมายถึงไม่ใช้ ImageNet pretrained weights
    model = models.resnet18(weights=None)

    model.fc = nn.Linear(
        model.fc.in_features,
        NUM_CLASSES
    )
    return model.to(device)

def build_training_objects(model):
    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=CFG["lr"],
        weight_decay=CFG["weight_decay"],
    )

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=CFG["epochs"],
    )

    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=(device.type == "cuda"),
    )

    return optimizer, scheduler, scaler

# ใช้กับ Multi-label Classification
criterion = nn.BCEWithLogitsLoss()

print("Model: ResNet-18 (Training from Scratch)")
print("Number of classes:", NUM_CLASSES)
print("Device:", device)
print("Loss:", criterion)

model = build_model()
dummy = torch.randn(
    2, 3, CFG["img_size"], CFG["img_size"],
    device=device,
)

model.eval()
with torch.no_grad():
    output = model(dummy)

print("Model device:", next(model.parameters()).device)
print("Input device:", dummy.device)
print("Output shape:", output.shape)

assert output.shape == (2, NUM_CLASSES)
assert next(model.parameters()).device == dummy.device

print("Model test passed!")
```

## 5. Checkpoint
แต่ละ checkpoint เก็บ **model + optimizer + scheduler + scaler + history + epoch** เพื่อให้เทรนต่อได้เหมือนไม่เคยหยุด
* `checkpoints/<การทดลอง>/last.pt` บันทึกทุก epoch (ใช้ resume เมื่อ session หลุดหรือหมดเวลา)
* `checkpoints/<การทดลอง>/best.pt` บันทึกเมื่อ validation loss ต่ำที่สุด (ใช้ประเมินผลและนำไปใช้งาน)

```python

# ==================================================
# 5. Checkpoint: Save and Resume Training
# ==================================================

import os
import torch


def ckpt_paths(exp):
    folder = os.path.join(CKPT_DIR, exp)
    os.makedirs(folder, exist_ok=True)

    last_path = os.path.join(folder, "last.pt")
    best_path = os.path.join(folder, "best.pt")

    return last_path, best_path


def save_checkpoint(
    path,
    epoch,
    model,
    optimizer,
    scheduler,
    scaler,
    history,
    best_val_loss,
    bad_epochs,
    stopped_early=False
):
    checkpoint = {
        "epoch": epoch,
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
        "scaler_state": scaler.state_dict(),
        "history": history,
        "best_val_loss": best_val_loss,
        "bad_epochs": bad_epochs,
        "stopped_early": stopped_early,
        "class_names": class_names,
        "cfg": CFG
    }

    torch.save(checkpoint, path)


def load_checkpoint(
    path,
    model,
    optimizer=None,
    scheduler=None,
    scaler=None
):
    checkpoint = torch.load(
        path,
        map_location=device,
        weights_only=False
    )

    # โหลดน้ำหนักเข้าโมเดล
    model.load_state_dict(checkpoint["model_state"])

    # ยืนยันให้โมเดลอยู่บน device ที่กำหนด
    model.to(device)

    if optimizer is not None:
        optimizer.load_state_dict(
            checkpoint["optimizer_state"]
        )

    if scheduler is not None:
        scheduler.load_state_dict(
            checkpoint["scheduler_state"]
        )

    if scaler is not None:
        scaler.load_state_dict(
            checkpoint["scaler_state"]
        )

    return checkpoint


print("Checkpoint directory:", CKPT_DIR)
print("Device:", device)
print("Save/Resume functions are ready.")
```

## 6. Training loop (resume อัตโนมัติถ้ามี `last.pt`)

```python
from sklearn.metrics import f1_score

def run_epoch(model, loader, training, optimizer=None, scaler=None):
    model.train(training)

    total_loss = 0.0
    total_exact = 0
    total_samples = 0
    all_true, all_prob = [], []

    context = torch.enable_grad() if training else torch.no_grad()

    with context:
        for images, targets in loader:
            images = images.to(device, non_blocking=True)
            targets = targets.to(device, non_blocking=True)

            if training:
                optimizer.zero_grad(set_to_none=True)

            with torch.amp.autocast(
                device_type="cuda",
                enabled=(device.type == "cuda")
            ):
                logits = model(images)
                loss = criterion(logits, targets)

            if training:
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()

            probs = torch.sigmoid(logits.detach())
            preds = (probs >= 0.5).float()

            n = images.size(0)
            total_loss += loss.item() * n
            total_exact += (preds == targets).all(dim=1).sum().item()
            total_samples += n

            all_true.append(targets.detach().cpu().numpy())
            all_prob.append(probs.detach().cpu().numpy())

    return (
        total_loss / max(total_samples, 1),
        total_exact / max(total_samples, 1),
        np.concatenate(all_true),
        np.concatenate(all_prob)
    )


histories = {}


def train_model(exp, until_epoch=None, force_restart=False):
    last_path, best_path = ckpt_paths(exp)

    # สำคัญ: สร้างโมเดลบน GPU และไม่ดาวน์โหลด weights
    model = build_model(pretrained=False).to(device)

    optimizer, scheduler, scaler = build_training_objects(model)

    history = {
        "train_loss": [],
        "val_loss": [],
        "train_acc": [],
        "val_acc": [],
        "train_f1": [],
        "val_f1": []
    }

    start_epoch = 1
    best_val_loss = float("inf")
    bad_epochs = 0

    # เริ่มใหม่โดยลบ checkpoint ของ experiment นี้
    if force_restart:
        for path in [last_path, best_path]:
            if os.path.exists(path):
                os.remove(path)

    # Resume จาก last.pt หากมี
    if os.path.exists(last_path):
        ckpt = load_checkpoint(
            last_path,
            model,
            optimizer,
            scheduler,
            scaler
        )

        history = ckpt["history"]
        start_epoch = int(ckpt["epoch"]) + 1
        best_val_loss = float(ckpt["best_val_loss"])
        bad_epochs = int(ckpt["bad_epochs"])

        print(
            f"Resumed {exp} from epoch {ckpt['epoch']}"
        )

        if ckpt.get("stopped_early", False):
            print("Checkpoint ระบุว่า Early Stopping ทำงานแล้ว")
            histories[exp] = history
            return history

    loader = make_train_loader(exp)

    end_epoch = (
        min(CFG["epochs"], until_epoch)
        if until_epoch is not None
        else CFG["epochs"]
    )

    for epoch in range(start_epoch, end_epoch + 1):

        train_loss, train_acc, ytr, ptr = run_epoch(
            model, loader, True, optimizer, scaler
        )

        val_loss, val_acc, yv, pv = run_epoch(
            model, val_loader, False
        )

        train_f1 = f1_score(
            ytr, ptr >= 0.5,
            average="macro",
            zero_division=0
        )

        val_f1 = f1_score(
            yv, pv >= 0.5,
            average="macro",
            zero_division=0
        )

        scheduler.step()

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["train_acc"].append(train_acc)
        history["val_acc"].append(val_acc)
        history["train_f1"].append(train_f1)
        history["val_f1"].append(val_f1)

        improved = val_loss < best_val_loss

        if improved:
            best_val_loss = val_loss
            bad_epochs = 0

            save_checkpoint(
                best_path, epoch, model,
                optimizer, scheduler, scaler,
                history, best_val_loss, bad_epochs,
                stopped_early=False
            )
        else:
            bad_epochs += 1

        should_stop = bad_epochs >= CFG["patience"]

        # บันทึกสถานะล่าสุดทุก epoch
        save_checkpoint(
            last_path, epoch, model,
            optimizer, scheduler, scaler,
            history, best_val_loss, bad_epochs,
            stopped_early=should_stop
        )

        print(
            f"{exp} | Epoch {epoch:02d}/{CFG['epochs']} "
            f"| train_loss={train_loss:.4f} "
            f"| val_loss={val_loss:.4f} "
            f"| train_exact_acc={train_acc:.4f} "
            f"| val_exact_acc={val_acc:.4f} "
            f"| val_macro_f1={val_f1:.4f}"
            + (" | BEST" if improved else "")
        )

        if should_stop:
            print(
                f"Early stopping: validation loss "
                f"ไม่ดีขึ้น {CFG['patience']} epochs"
            )
            break

    if (
        until_epoch is not None
        and end_epoch < CFG["epochs"]
        and bad_epochs < CFG["patience"]
    ):
        print(f"หยุดชั่วคราวที่ epoch {end_epoch} เพื่อสาธิต Resume")

    histories[exp] = history
    return history
```

## 9. อภิปรายผล
1. **Convergence:** แต่ละการทดลอง converge ที่ epoch เท่าไร ดูจากอะไร early stopping หยุดเองหรือไม่
   ...
2. **Overfitting:** ในการทดลอง A ช่องว่างระหว่าง train loss กับ val loss เป็นอย่างไรเมื่อเทียบกับ B
   ...
3. **ผลของ augmentation:** augmentation ทำให้ val loss / val accuracy และรูปร่างของ loss curve เปลี่ยนไปอย่างไร (เช่น converge ช้าลง, train loss สูงขึ้น) เพราะอะไร
   ...
4. **Test performance:** accuracy / macro F1 ของโมเดลสุดท้ายเท่าไร class ไหนทำได้ดีและแย่ที่สุด เพราะอะไร
   ...
5. **Error analysis:** โมเดลสับสน class ไหนกับ class ไหนบ่อย ภาพที่ผิดมีลักษณะร่วมกันอย่างไร
   ...
6. **แนวทางปรับปรุง:** ถ้ามีเวลาเพิ่มจะทดลองอะไรต่อ เพราะอะไร
   ...
