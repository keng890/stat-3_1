## 2. Data Loading และ Data Preparation
* แบ่งข้อมูลแบบ **stratified** 70/15/15 เพื่อให้สัดส่วนแต่ละ class เท่ากันทุกชุด
* validation/test ใช้แค่ resize + normalize เสมอ (augment เฉพาะ train set)

```python

from pathlib import Path
from collections import Counter

import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from torchvision import transforms
import matplotlib.pyplot as plt
ROOT = Path(
    "/kaggle/input/datasets/pkdarabi/"
    "bone-fracture-detection-computer-vision-project/"
    "BoneFractureYolo8"
)

# ตรวจว่าพาธมีอยู่จริง
assert ROOT.exists(), f"ไม่พบ Dataset: {ROOT}"

# ชื่อคลาสจาก data.yaml
class_names = [
    "elbow positive",
    "fingers positive",
    "forearm fracture",
    "humerus fracture",
    "humerus",
    "shoulder fracture",
    "wrist positive",
]

NUM_CLASSES = len(class_names)
S = CFG["img_size"]

# Transform สำหรับ validation/test
eval_tf = transforms.Compose([
    transforms.Resize((S, S)),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    ),
])

class BoneFractureDataset(Dataset):
    def __init__(self, root, split, transform=None):
        self.image_dir = Path(root) / split / "images"
        self.label_dir = Path(root) / split / "labels"
        self.transform = transform

        extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
        self.images = sorted([
            p for p in self.image_dir.iterdir()
            if p.suffix.lower() in extensions
        ])

        if not self.images:
            raise RuntimeError(
                f"ไม่พบภาพใน {self.image_dir}"
            )

    def __len__(self):
        return len(self.images)

    def __getitem__(self, index):
        image_path = self.images[index]

        # โหลดภาพ
        image = Image.open(image_path).convert("RGB")

        # Multi-hot label: หนึ่งภาพอาจมีหลายคลาส
        target = torch.zeros(NUM_CLASSES, dtype=torch.float32)
        label_path = self.label_dir / f"{image_path.stem}.txt"

        if label_path.exists():
            for line in label_path.read_text().splitlines():
                parts = line.split()
                if not parts:
                    continue

                class_id = int(parts[0])
                if 0 <= class_id < NUM_CLASSES:
                    target[class_id] = 1.0

        if self.transform:
            image = self.transform(image)

        return image, target

train_ds = BoneFractureDataset(ROOT, "train", eval_tf)
val_ds = BoneFractureDataset(ROOT, "valid", eval_tf)
test_ds = BoneFractureDataset(ROOT, "test", eval_tf)

print("จำนวนภาพ")
print("Train:", len(train_ds))
print("Validation:", len(val_ds))
print("Test:", len(test_ds))
print("Classes:", class_names)
```

### สำรวจข้อมูล: การกระจายของ class และตัวอย่างภาพ

```python

# ตรวจการกระจายคลาสจาก Label
for split in ["train", "valid", "test"]:
    label_dir = ROOT / split / "labels"
    class_counts = Counter()

    for path in label_dir.glob("*.txt"):
        for line in path.read_text().splitlines():
            parts = line.split()
            if parts:
                class_id = int(parts[0])
                if 0 <= class_id < NUM_CLASSES:
                    class_counts[class_id] += 1

    print(f"\n{split.upper()}")
    for class_id, name in enumerate(class_names):
        print(f"{name}: {class_counts[class_id]}")

# แสดงภาพตัวอย่าง
fig, axes = plt.subplots(2, 8, figsize=(16, 5))

for ax, i in zip(axes.ravel(), np.random.choice(
    len(train_ds), min(16, len(train_ds)), replace=False
)):
    image, target = train_ds[i]
    mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
    std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
    shown = (image * std + mean).clamp(0, 1)

    ax.imshow(shown.permute(1, 2, 0))
    present = [
        class_names[j] for j in range(NUM_CLASSES)
        if target[j] > 0
    ]
    ax.set_title("\n".join(present) if present else "No label",
                 fontsize=7)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## 3. Data Augmentation (กลุ่มออกแบบเอง)
ใส่ augmentation ที่ **สมเหตุสมผลกับภาพของกลุ่ม** แล้วอธิบายเหตุผลในตาราง
ดูตัวเลือกได้ที่ [torchvision transforms](https://pytorch.org/vision/stable/transforms.html)

| Augmentation | ค่าที่ใช้ | เหตุผล (ภาพจริงมีความแปรผันแบบนี้หรือไม่) |
|---|---|---|
| ... | ... | ... |

```python

from torchvision import transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import numpy as np
import torch

S = CFG["img_size"]

MEAN = [0.485, 0.456, 0.406]
STD  = [0.229, 0.224, 0.225]

# A: ไม่มี augmentation
eval_tf = transforms.Compose([
    transforms.Resize((S, S)),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

# B: มี augmentation
aug_tf = transforms.Compose([
    transforms.Resize((S, S)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=5),
    transforms.RandomAffine(
        degrees=0,
        translate=(0.03, 0.03),
        scale=(0.95, 1.05),
    ),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

# สร้าง Dataset แยกกัน โดยใช้ภาพ Train ชุดเดียวกัน
train_ds = {
    "A_no_aug": BoneFractureDataset(
        ROOT, "train", transform=eval_tf
    ),
    "B_aug": BoneFractureDataset(
        ROOT, "train", transform=aug_tf
    ),
}

# Validation และ Test ไม่ทำ augmentation
val_ds = BoneFractureDataset(
    ROOT, "valid", transform=eval_tf
)
test_ds = BoneFractureDataset(
    ROOT, "test", transform=eval_tf
)

print("Train A:", len(train_ds["A_no_aug"]))
print("Train B:", len(train_ds["B_aug"]))
print("Validation:", len(val_ds))
print("Test:", len(test_ds))
```

```python

def denorm(x):
    mean = torch.tensor(MEAN).view(3, 1, 1)
    std = torch.tensor(STD).view(3, 1, 1)
    return (x.cpu() * std + mean).clamp(0, 1).permute(1, 2, 0).numpy()

original_ds = train_ds["A_no_aug"]
augmented_ds = train_ds["B_aug"]

rng = np.random.default_rng(SEED)
sample_ids = rng.choice(
    len(original_ds),
    size=min(3, len(original_ds)),
    replace=False
)

fig, axes = plt.subplots(3, 8, figsize=(16, 6.5))

for row_idx, image_idx in enumerate(sample_ids):
    # ภาพต้นฉบับ
    image, target = original_ds[image_idx]
    axes[row_idx, 0].imshow(denorm(image), cmap="gray")
    axes[row_idx, 0].set_title("Original", fontsize=9)

    # สุ่ม augmentation 7 ครั้งจากภาพเดิม
    for col_idx in range(1, 8):
        aug_image, _ = augmented_ds[image_idx]
        axes[row_idx, col_idx].imshow(
            denorm(aug_image), cmap="gray"
        )
        axes[row_idx, col_idx].set_title(
            f"Aug {col_idx}", fontsize=9
        )

    for ax in axes[row_idx]:
        ax.axis("off")

plt.tight_layout()
plt.show()
```

## 4. Model: Transfer Learning ด้วย ResNet-18
ใช้ pretrained weights จาก ImageNet แล้วเปลี่ยนชั้นสุดท้ายให้เท่ากับจำนวน class
(กลุ่มสามารถเปลี่ยนเป็น EfficientNet, MobileNet, ConvNeXt ฯลฯ ได้ แต่ต้องใช้โมเดลเดียวกันในการทดลอง A และ B)

```python

import torch
import torch.nn as nn
from torchvision import models

# --------------------------------------------------
# 1. Build Model
# --------------------------------------------------

def build_model(pretrained=False):
    # ไม่ดาวน์โหลด pretrained weights จากอินเทอร์เน็ต
    model = models.resnet18(weights=None)

    # เปลี่ยน output layer ให้รองรับจำนวนคลาสของ dataset
    model.fc = nn.Linear(
        model.fc.in_features,
        NUM_CLASSES
    )

    # ย้ายโมเดลไปยัง CPU หรือ GPU ตาม device ที่กำหนด
    model = model.to(device)

    return model


# --------------------------------------------------
# 2. Build Training Objects
# --------------------------------------------------

def build_training_objects(model):
    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=CFG["lr"],
        weight_decay=CFG["weight_decay"]
    )

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=CFG["epochs"]
    )

    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=(device.type == "cuda")
    )

    return optimizer, scheduler, scaler


# --------------------------------------------------
# 3. Loss Function
# --------------------------------------------------

# Dataset เป็น multi-label classification
criterion = nn.BCEWithLogitsLoss()


# --------------------------------------------------
# 4. Print Configuration
# --------------------------------------------------

print("สร้าง Model และ Training Objects พร้อมแล้ว")
print("จำนวนคลาส:", NUM_CLASSES)
print("Device:", device)
print("Loss:", criterion)


# --------------------------------------------------
# 5. Test Model
# --------------------------------------------------

model = build_model(pretrained=False)

dummy = torch.randn(
    2,
    3,
    CFG["img_size"],
    CFG["img_size"]
).to(device)

model.eval()

with torch.no_grad():
    output = model(dummy)

print("Model device:", next(model.parameters()).device)
print("Input device:", dummy.device)
print("Output shape:", output.shape)

assert output.shape == (2, NUM_CLASSES), (
    f"Expected output shape (2, {NUM_CLASSES}), "
    f"but got {tuple(output.shape)}"
)

assert next(model.parameters()).device == dummy.device, (
    "Model and input are on different devices!"
)

print("Model test passed!")
```

## 5. Checkpoint
แต่ละ checkpoint เก็บ **model + optimizer + scheduler + scaler + history + epoch** เพื่อให้เทรนต่อได้เหมือนไม่เคยหยุด
* `checkpoints/<การทดลอง>/last.pt` บันทึกทุก epoch (ใช้ resume เมื่อ session หลุดหรือหมดเวลา)
* `checkpoints/<การทดลอง>/best.pt` บันทึกเมื่อ validation loss ต่ำที่สุด (ใช้ประเมินผลและนำไปใช้งาน)

```python

# ==================================================
# 5. Checkpoint: Save and Resume Training
# ==================================================

import os
import torch


def ckpt_paths(exp):
    folder = os.path.join(CKPT_DIR, exp)
    os.makedirs(folder, exist_ok=True)

    last_path = os.path.join(folder, "last.pt")
    best_path = os.path.join(folder, "best.pt")

    return last_path, best_path


def save_checkpoint(
    path,
    epoch,
    model,
    optimizer,
    scheduler,
    scaler,
    history,
    best_val_loss,
    bad_epochs,
    stopped_early=False
):
    checkpoint = {
        "epoch": epoch,
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
        "scaler_state": scaler.state_dict(),
        "history": history,
        "best_val_loss": best_val_loss,
        "bad_epochs": bad_epochs,
        "stopped_early": stopped_early,
        "class_names": class_names,
        "cfg": CFG
    }

    torch.save(checkpoint, path)


def load_checkpoint(
    path,
    model,
    optimizer=None,
    scheduler=None,
    scaler=None
):
    checkpoint = torch.load(
        path,
        map_location=device,
        weights_only=False
    )

    # โหลดน้ำหนักเข้าโมเดล
    model.load_state_dict(checkpoint["model_state"])

    # ยืนยันให้โมเดลอยู่บน device ที่กำหนด
    model.to(device)

    if optimizer is not None:
        optimizer.load_state_dict(
            checkpoint["optimizer_state"]
        )

    if scheduler is not None:
        scheduler.load_state_dict(
            checkpoint["scheduler_state"]
        )

    if scaler is not None:
        scaler.load_state_dict(
            checkpoint["scaler_state"]
        )

    return checkpoint


print("Checkpoint directory:", CKPT_DIR)
print("Device:", device)
print("Save/Resume functions are ready.")
```

## 6. Training loop (resume อัตโนมัติถ้ามี `last.pt`)

```python
from sklearn.metrics import f1_score

def run_epoch(model, loader, training, optimizer=None, scaler=None):
    model.train(training)

    total_loss = 0.0
    total_exact = 0
    total_samples = 0
    all_true, all_prob = [], []

    context = torch.enable_grad() if training else torch.no_grad()

    with context:
        for images, targets in loader:
            images = images.to(device, non_blocking=True)
            targets = targets.to(device, non_blocking=True)

            if training:
                optimizer.zero_grad(set_to_none=True)

            with torch.amp.autocast(
                device_type="cuda",
                enabled=(device.type == "cuda")
            ):
                logits = model(images)
                loss = criterion(logits, targets)

            if training:
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()

            probs = torch.sigmoid(logits.detach())
            preds = (probs >= 0.5).float()

            n = images.size(0)
            total_loss += loss.item() * n
            total_exact += (preds == targets).all(dim=1).sum().item()
            total_samples += n

            all_true.append(targets.detach().cpu().numpy())
            all_prob.append(probs.detach().cpu().numpy())

    return (
        total_loss / max(total_samples, 1),
        total_exact / max(total_samples, 1),
        np.concatenate(all_true),
        np.concatenate(all_prob)
    )


histories = {}


def train_model(exp, until_epoch=None, force_restart=False):
    last_path, best_path = ckpt_paths(exp)

    # สำคัญ: สร้างโมเดลบน GPU และไม่ดาวน์โหลด weights
    model = build_model(pretrained=False).to(device)

    optimizer, scheduler, scaler = build_training_objects(model)

    history = {
        "train_loss": [],
        "val_loss": [],
        "train_acc": [],
        "val_acc": [],
        "train_f1": [],
        "val_f1": []
    }

    start_epoch = 1
    best_val_loss = float("inf")
    bad_epochs = 0

    # เริ่มใหม่โดยลบ checkpoint ของ experiment นี้
    if force_restart:
        for path in [last_path, best_path]:
            if os.path.exists(path):
                os.remove(path)

    # Resume จาก last.pt หากมี
    if os.path.exists(last_path):
        ckpt = load_checkpoint(
            last_path,
            model,
            optimizer,
            scheduler,
            scaler
        )

        history = ckpt["history"]
        start_epoch = int(ckpt["epoch"]) + 1
        best_val_loss = float(ckpt["best_val_loss"])
        bad_epochs = int(ckpt["bad_epochs"])

        print(
            f"Resumed {exp} from epoch {ckpt['epoch']}"
        )

        if ckpt.get("stopped_early", False):
            print("Checkpoint ระบุว่า Early Stopping ทำงานแล้ว")
            histories[exp] = history
            return history

    loader = make_train_loader(exp)

    end_epoch = (
        min(CFG["epochs"], until_epoch)
        if until_epoch is not None
        else CFG["epochs"]
    )

    for epoch in range(start_epoch, end_epoch + 1):

        train_loss, train_acc, ytr, ptr = run_epoch(
            model, loader, True, optimizer, scaler
        )

        val_loss, val_acc, yv, pv = run_epoch(
            model, val_loader, False
        )

        train_f1 = f1_score(
            ytr, ptr >= 0.5,
            average="macro",
            zero_division=0
        )

        val_f1 = f1_score(
            yv, pv >= 0.5,
            average="macro",
            zero_division=0
        )

        scheduler.step()

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["train_acc"].append(train_acc)
        history["val_acc"].append(val_acc)
        history["train_f1"].append(train_f1)
        history["val_f1"].append(val_f1)

        improved = val_loss < best_val_loss

        if improved:
            best_val_loss = val_loss
            bad_epochs = 0

            save_checkpoint(
                best_path, epoch, model,
                optimizer, scheduler, scaler,
                history, best_val_loss, bad_epochs,
                stopped_early=False
            )
        else:
            bad_epochs += 1

        should_stop = bad_epochs >= CFG["patience"]

        # บันทึกสถานะล่าสุดทุก epoch
        save_checkpoint(
            last_path, epoch, model,
            optimizer, scheduler, scaler,
            history, best_val_loss, bad_epochs,
            stopped_early=should_stop
        )

        print(
            f"{exp} | Epoch {epoch:02d}/{CFG['epochs']} "
            f"| train_loss={train_loss:.4f} "
            f"| val_loss={val_loss:.4f} "
            f"| train_exact_acc={train_acc:.4f} "
            f"| val_exact_acc={val_acc:.4f} "
            f"| val_macro_f1={val_f1:.4f}"
            + (" | BEST" if improved else "")
        )

        if should_stop:
            print(
                f"Early stopping: validation loss "
                f"ไม่ดีขึ้น {CFG['patience']} epochs"
            )
            break

    if (
        until_epoch is not None
        and end_epoch < CFG["epochs"]
        and bad_epochs < CFG["patience"]
    ):
        print(f"หยุดชั่วคราวที่ epoch {end_epoch} เพื่อสาธิต Resume")

    histories[exp] = history
    return history
```

## 8. Model Evaluation บน Test Set (โหลด `best.pt` ของโมเดลสุดท้าย, ประเมินครั้งเดียว)

```python
#8.1
from sklearn.metrics import (
    classification_report,
    accuracy_score,
    precision_recall_fscore_support,
    multilabel_confusion_matrix
)

if "FINAL_EXP" not in globals():
    raise RuntimeError("กรุณารันข้อ 7 เพื่อเลือก FINAL_EXP ก่อน")

best_path = ckpt_paths(FINAL_EXP)[1]

if not os.path.exists(best_path):
    raise FileNotFoundError(
        f"ไม่พบ checkpoint: {best_path} กรุณารันข้อ 6 ก่อน"
    )

# โหลด best model
best_model = build_model(pretrained=False).to(device)
ckpt = load_checkpoint(best_path, best_model)
best_model.eval()

print(
    f"Loaded {FINAL_EXP}/best.pt | "
    f"Best epoch={ckpt['epoch']} | "
    f"Best val loss={ckpt['best_val_loss']:.4f}"
)

all_true, all_prob = [], []

with torch.no_grad():
    for x, y in test_loader:
        x = x.to(device, non_blocking=True)

        with torch.amp.autocast(
            device_type="cuda",
            enabled=(device.type == "cuda")
        ):
            logits = best_model(x)

        all_prob.append(
            torch.sigmoid(logits).cpu().numpy()
        )
        all_true.append(y.numpy())

y_true = np.concatenate(all_true).astype(int)
y_prob = np.concatenate(all_prob)
y_pred = (y_prob >= 0.5).astype(int)

# Exact-match accuracy:
# ทุกคลาสของภาพนั้นต้องทำนายถูกทั้งหมดจึงนับว่าถูก
exact_acc = accuracy_score(y_true, y_pred)

prec_micro, rec_micro, f1_micro, _ = (
    precision_recall_fscore_support(
        y_true, y_pred,
        average="micro",
        zero_division=0
    )
)

prec_macro, rec_macro, f1_macro, _ = (
    precision_recall_fscore_support(
        y_true, y_pred,
        average="macro",
        zero_division=0
    )
)

print(f"Exact-match accuracy: {exact_acc:.4f}")

print(
    f"Micro Precision={prec_micro:.4f} | "
    f"Micro Recall={rec_micro:.4f} | "
    f"Micro F1={f1_micro:.4f}"
)

print(
    f"Macro Precision={prec_macro:.4f} | "
    f"Macro Recall={rec_macro:.4f} | "
    f"Macro F1={f1_macro:.4f}"
)

print("\nPer-class classification report:")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        zero_division=0,
        digits=3
    )
)

report = classification_report(
    y_true,
    y_pred,
    target_names=class_names,
    zero_division=0,
    output_dict=True
)

metrics = {
    "final_experiment": FINAL_EXP,
    "best_epoch": int(ckpt["epoch"]),
    "test_exact_match_accuracy": float(exact_acc),
    "micro_precision": float(prec_micro),
    "micro_recall": float(rec_micro),
    "micro_f1": float(f1_micro),
    "macro_precision": float(prec_macro),
    "macro_recall": float(rec_macro),
    "macro_f1": float(f1_macro),
    "report": report
}

with open(
    os.path.join(WORK_DIR, "test_metrics.json"),
    "w"
) as f:
    json.dump(metrics, f, indent=2)
```
