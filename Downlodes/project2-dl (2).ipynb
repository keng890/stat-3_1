8. **เลือกโมเดลสุดท้ายจาก validation loss** แล้วประเมินบน test set **ครั้งเดียว**: accuracy, precision/recall/F1 ราย class, confusion matrix และตัวอย่างภาพที่ทำนายผิด

## 2. Data Loading และ Data Preparation
* แบ่งข้อมูลแบบ **stratified** 70/15/15 เพื่อให้สัดส่วนแต่ละ class เท่ากันทุกชุด
* validation/test ใช้แค่ resize + normalize เสมอ (augment เฉพาะ train set)

```python

from pathlib import Path
from collections import Counter

import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from torchvision import transforms
import matplotlib.pyplot as plt
ROOT = Path(
    "/kaggle/input/datasets/pkdarabi/"
    "bone-fracture-detection-computer-vision-project/"
    "BoneFractureYolo8"
)

# ตรวจว่าพาธมีอยู่จริง
assert ROOT.exists(), f"ไม่พบ Dataset: {ROOT}"

# ชื่อคลาสจาก data.yaml
class_names = [
    "elbow positive",
    "fingers positive",
    "forearm fracture",
    "humerus fracture",
    "humerus",
    "shoulder fracture",
    "wrist positive",
]

NUM_CLASSES = len(class_names)
S = CFG["img_size"]

# Transform สำหรับ validation/test
eval_tf = transforms.Compose([
    transforms.Resize((S, S)),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    ),
])

class BoneFractureDataset(Dataset):
    def __init__(self, root, split, transform=None):
        self.image_dir = Path(root) / split / "images"
        self.label_dir = Path(root) / split / "labels"
        self.transform = transform

        extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
        self.images = sorted([
            p for p in self.image_dir.iterdir()
            if p.suffix.lower() in extensions
        ])

        if not self.images:
            raise RuntimeError(
                f"ไม่พบภาพใน {self.image_dir}"
            )

    def __len__(self):
        return len(self.images)

    def __getitem__(self, index):
        image_path = self.images[index]

        # โหลดภาพ
        image = Image.open(image_path).convert("RGB")

        # Multi-hot label: หนึ่งภาพอาจมีหลายคลาส
        target = torch.zeros(NUM_CLASSES, dtype=torch.float32)
        label_path = self.label_dir / f"{image_path.stem}.txt"

        if label_path.exists():
            for line in label_path.read_text().splitlines():
                parts = line.split()
                if not parts:
                    continue

                class_id = int(parts[0])
                if 0 <= class_id < NUM_CLASSES:
                    target[class_id] = 1.0

        if self.transform:
            image = self.transform(image)

        return image, target

train_ds = BoneFractureDataset(ROOT, "train", eval_tf)
val_ds = BoneFractureDataset(ROOT, "valid", eval_tf)
test_ds = BoneFractureDataset(ROOT, "test", eval_tf)

print("จำนวนภาพ")
print("Train:", len(train_ds))
print("Validation:", len(val_ds))
print("Test:", len(test_ds))
print("Classes:", class_names)
```

### สำรวจข้อมูล: การกระจายของ class และตัวอย่างภาพ

```python

# ตรวจการกระจายคลาสจาก Label
for split in ["train", "valid", "test"]:
    label_dir = ROOT / split / "labels"
    class_counts = Counter()

    for path in label_dir.glob("*.txt"):
        for line in path.read_text().splitlines():
            parts = line.split()
            if parts:
                class_id = int(parts[0])
                if 0 <= class_id < NUM_CLASSES:
                    class_counts[class_id] += 1

    print(f"\n{split.upper()}")
    for class_id, name in enumerate(class_names):
        print(f"{name}: {class_counts[class_id]}")

# แสดงภาพตัวอย่าง
fig, axes = plt.subplots(2, 8, figsize=(16, 5))

for ax, i in zip(axes.ravel(), np.random.choice(
    len(train_ds), min(16, len(train_ds)), replace=False
)):
    image, target = train_ds[i]
    mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
    std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
    shown = (image * std + mean).clamp(0, 1)

    ax.imshow(shown.permute(1, 2, 0))
    present = [
        class_names[j] for j in range(NUM_CLASSES)
        if target[j] > 0
    ]
    ax.set_title("\n".join(present) if present else "No label",
                 fontsize=7)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## 3. Data Augmentation (กลุ่มออกแบบเอง)
ใส่ augmentation ที่ **สมเหตุสมผลกับภาพของกลุ่ม** แล้วอธิบายเหตุผลในตาราง
ดูตัวเลือกได้ที่ [torchvision transforms](https://pytorch.org/vision/stable/transforms.html)

| Augmentation | ค่าที่ใช้ | เหตุผล (ภาพจริงมีความแปรผันแบบนี้หรือไม่) |
|---|---|---|
| ... | ... | ... |

```python

from torchvision import transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import numpy as np
import torch

S = CFG["img_size"]

MEAN = [0.485, 0.456, 0.406]
STD  = [0.229, 0.224, 0.225]

# A: ไม่มี augmentation
eval_tf = transforms.Compose([
    transforms.Resize((S, S)),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

# B: มี augmentation
aug_tf = transforms.Compose([
    transforms.Resize((S, S)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=5),
    transforms.RandomAffine(
        degrees=0,
        translate=(0.03, 0.03),
        scale=(0.95, 1.05),
    ),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

# สร้าง Dataset แยกกัน โดยใช้ภาพ Train ชุดเดียวกัน
train_ds = {
    "A_no_aug": BoneFractureDataset(
        ROOT, "train", transform=eval_tf
    ),
    "B_aug": BoneFractureDataset(
        ROOT, "train", transform=aug_tf
    ),
}

# Validation และ Test ไม่ทำ augmentation
val_ds = BoneFractureDataset(
    ROOT, "valid", transform=eval_tf
)
test_ds = BoneFractureDataset(
    ROOT, "test", transform=eval_tf
)

print("Train A:", len(train_ds["A_no_aug"]))
print("Train B:", len(train_ds["B_aug"]))
print("Validation:", len(val_ds))
print("Test:", len(test_ds))
```

```python

def denorm(x):
    mean = torch.tensor(MEAN).view(3, 1, 1)
    std = torch.tensor(STD).view(3, 1, 1)
    return (x.cpu() * std + mean).clamp(0, 1).permute(1, 2, 0).numpy()

original_ds = train_ds["A_no_aug"]
augmented_ds = train_ds["B_aug"]

rng = np.random.default_rng(SEED)
sample_ids = rng.choice(
    len(original_ds),
    size=min(3, len(original_ds)),
    replace=False
)

fig, axes = plt.subplots(3, 8, figsize=(16, 6.5))

for row_idx, image_idx in enumerate(sample_ids):
    # ภาพต้นฉบับ
    image, target = original_ds[image_idx]
    axes[row_idx, 0].imshow(denorm(image), cmap="gray")
    axes[row_idx, 0].set_title("Original", fontsize=9)

    # สุ่ม augmentation 7 ครั้งจากภาพเดิม
    for col_idx in range(1, 8):
        aug_image, _ = augmented_ds[image_idx]
        axes[row_idx, col_idx].imshow(
            denorm(aug_image), cmap="gray"
        )
        axes[row_idx, col_idx].set_title(
            f"Aug {col_idx}", fontsize=9
        )

    for ax in axes[row_idx]:
        ax.axis("off")

plt.tight_layout()
plt.show()
```

```python

loader_kw = {
    "batch_size": CFG["batch_size"],
    "num_workers": CFG["num_workers"],
    "pin_memory": torch.cuda.is_available(),
}

val_loader = DataLoader(
    val_ds,
    shuffle=False,
    **loader_kw
)

test_loader = DataLoader(
    test_ds,
    shuffle=False,
    **loader_kw
)

def make_train_loader(exp):
    set_seed()
    return DataLoader(
        train_ds[exp],
        shuffle=True,
        **loader_kw
    )

train_loader_A = make_train_loader("A_no_aug")
train_loader_B = make_train_loader("B_aug")

print("เตรียม DataLoader สำเร็จ")
print("Train A batches:", len(train_loader_A))
print("Train B batches:", len(train_loader_B))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

## 4. Model: Transfer Learning ด้วย ResNet-18
ใช้ pretrained weights จาก ImageNet แล้วเปลี่ยนชั้นสุดท้ายให้เท่ากับจำนวน class
(กลุ่มสามารถเปลี่ยนเป็น EfficientNet, MobileNet, ConvNeXt ฯลฯ ได้ แต่ต้องใช้โมเดลเดียวกันในการทดลอง A และ B)

```python

import torch
import torch.nn as nn
from torchvision import models

def build_model(pretrained=True):
    weights = models.ResNet18_Weights.DEFAULT if pretrained else None
    m = models.resnet18(weights=weights)

    # Output 7 ค่า สำหรับ 7 คลาส
    m.fc = nn.Linear(m.fc.in_features, NUM_CLASSES)

    return m.to(device)


def build_training_objects(model):
    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=CFG["lr"],
        weight_decay=CFG["weight_decay"]
    )

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=CFG["epochs"]
    )

    scaler = torch.amp.GradScaler("cuda", enabled=(device.type == "cuda"))

    return optimizer, scheduler, scaler


# Dataset ของเราเป็น multi-label
criterion = nn.BCEWithLogitsLoss()

print("สร้าง Model และ Training Objects พร้อมแล้ว")
print("จำนวนคลาส:", NUM_CLASSES)
print("Device:", device)
print("Loss:", criterion)


##---------------------------------------------------------

model = build_model(pretrained=False)

dummy = torch.randn(
    2, 3, CFG["img_size"], CFG["img_size"]
).to(device)

model.eval()
with torch.no_grad():
    output = model(dummy)

print("Output shape:", output.shape)
assert output.shape == (2, NUM_CLASSES)
print("Model test passed!")
```

## 5. Checkpoint
แต่ละ checkpoint เก็บ **model + optimizer + scheduler + scaler + history + epoch** เพื่อให้เทรนต่อได้เหมือนไม่เคยหยุด
* `checkpoints/<การทดลอง>/last.pt` บันทึกทุก epoch (ใช้ resume เมื่อ session หลุดหรือหมดเวลา)
* `checkpoints/<การทดลอง>/best.pt` บันทึกเมื่อ validation loss ต่ำที่สุด (ใช้ประเมินผลและนำไปใช้งาน)

```python
def ckpt_paths(exp):
    d = f"{CKPT_DIR}/{exp}"; os.makedirs(d, exist_ok=True)
    return f"{d}/last.pt", f"{d}/best.pt"

def save_checkpoint(path, epoch, model, optimizer, scheduler, scaler, history, best_val_loss, bad_epochs, stopped_early):
    torch.save({
        "epoch": epoch,
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
        "scaler_state": scaler.state_dict(),
        "history": history,
        "best_val_loss": best_val_loss,
        "bad_epochs": bad_epochs,
        "stopped_early": stopped_early,
        "class_names": class_names,
        "cfg": CFG,
    }, path)

def load_checkpoint(path, model, optimizer=None, scheduler=None, scaler=None):
    ckpt = torch.load(path, map_location=device, weights_only=False)
    model.load_state_dict(ckpt["model_state"])
    if optimizer is not None: optimizer.load_state_dict(ckpt["optimizer_state"])
    if scheduler is not None: scheduler.load_state_dict(ckpt["scheduler_state"])
    if scaler is not None:    scaler.load_state_dict(ckpt["scaler_state"])
    return ckpt
```

## 8. Model Evaluation บน Test Set (โหลด `best.pt` ของโมเดลสุดท้าย, ประเมินครั้งเดียว)

```python
best_model = build_model(pretrained=False)
ckpt = load_checkpoint(ckpt_paths(FINAL_EXP)[1], best_model)
best_model.eval()
print(f"Loaded {FINAL_EXP}/best.pt จาก epoch {ckpt['epoch']} (val loss {ckpt['best_val_loss']:.4f})")

y_true, y_pred = [], []
with torch.no_grad(), torch.amp.autocast("cuda"):
    for x, y in test_loader:
        y_pred.append(best_model(x.to(device)).argmax(1).cpu().numpy())
        y_true.append(y.numpy())
y_true, y_pred = np.concatenate(y_true), np.concatenate(y_pred)

test_acc = accuracy_score(y_true, y_pred)
print(f"Test accuracy: {test_acc:.4f}\n")
print(classification_report(y_true, y_pred, target_names=class_names, digits=3))

json.dump({"final_experiment": FINAL_EXP, "best_epoch": ckpt["epoch"], "test_accuracy": test_acc,
           "report": classification_report(y_true, y_pred, target_names=class_names, output_dict=True)},
          open(f"{WORK_DIR}/test_metrics.json", "w"), indent=2)
```

   ...

   ...

   ...
