## 1. ปัญหาและคำถามที่สนใจ
**ปัญหา/คำถามที่สนใจคืออะไร**

สไตล์การบุกและการเคลื่อนที่ทำทาง (Movement Archetypes) แตกต่างกันอย่างไร และทีมที่มี win rate สูงมีสไตล์การเล่นอย่างไร

**ปัญหา/คำถามนี้มีความสำคัญอย่างไร และใครจะนำผลไปใช้**

- ความสำคัญ: ช่วยเปลี่ยนข้อมูลพิกัดการเคลื่อนที่ดิบปริมาณมหาศาล (Big Data) ให้กลายเป็น "ยุทธวิธีเชิงพื้นที่ (Tactical Insights)" ทำให้เห็นรูปแบบการบุกที่เป็นแม่แบบสากล และเข้าใจโครงสร้างเกมบุกของทีมที่ประสบความสำเร็จเพื่อวางแนวทางตั้งรับดักทางได้อย่างตรงจุด

- ผู้นำผลไปใช้:  
  -โค้ชและทีมงานสเกาต์ (Coaches & Tactical Analysts): นำโปรไฟล์สไตล์การเล่นของทีมชั้นนำไปศึกษาวิเคราะห์เพื่อวางแผนการซ้อมและจัดแท็กติกรับมือคู่แข่งก่อนลงแข่งจริง

  -นักกีฬา (Players): นำภาพจำลองเส้นทางการเคลื่อนที่ (2D Pitch Map) ไปศึกษาเพื่ออ่านเกมและยืนปิดช่องทางส่ง/เลี้ยงบอลล่วงหน้า

   -นักวิเคราะห์ข้อมูลสโมสร (Sports Data Scientists): นำ Pipeline การจัดกลุ่มนี้ไปใช้ประเมินประสิทธิภาพและเปรียบเทียบสไตล์การเล่นของนักเตะและทีมสโมสร

**เป้าหมาย (target) และประเภทงาน (classification / regression / unsupervised)**

เป้าหมาย (Target):

Target ระดับโมเดล: กลุ่มรูปแบบการเคลื่อนที่ทำทาง (movement_archetype หรือ Cluster ID) ที่สกัดจากฟีเจอร์เชิงพื้นที่ ($\Delta X, \Delta Y, \text{Distance}, \text{Angle}$)

Target ระดับแท็กติก: สัดส่วนเปอร์เซ็นต์การเลือกใช้ Movement Archetypes แต่ละรูปแบบของทีมที่มีอัตราการชนะสูง (High Win-Rate Teams)

ประเภทงาน (Task Type): Unsupervised Learning (การจัดกลุ่มด้วย MiniBatchKMeans / KMeans) ร่วมกับ Descriptive Analytics (การสกัดโปรไฟล์รายทีมด้วย Cross-tabulation)

## 2. Data Description
* แหล่งข้อมูล (ลิงก์) และช่วงเวลาของข้อมูล

  แหล่งข้อมูล: [StatsBomb Open Data Repository (GitHub Raw JSON)](https://github.com/hudl/open-data)

  ช่วงเวลาของข้อมูล: ครอบคลุมการแข่งขันฟุตบอลทัวร์นาเมนต์หลักระดับโลกตั้งแต่ปี 2004 – 2024 (เช่น FIFA World Cup, UEFA Euro, UEFA Champions League)

---

* จำนวนแถว จำนวนคอลัมน์ และความหมายของแต่ละคอลัมน์ที่ใช้ (data dictionary)

  ประมวลผลโดยนำตารางดิบ 2 ตาราง มาเชื่อมโยงกันด้วยคอลัมน์ match_id (Inner Join):
1. ตาราง matches  ใช้ดึงผลการแข่งขันเพื่อคำนวณอัตราการชนะ (Win Rate)
2. ตาราง events ใช้ดึงพิกัดการเคลื่อนที่และประเภทเหตุการณ์

* จำนวนแถว (Rows): ประมาณ 9M แถว (เหตุการณ์ทั้งหมดจากทุกแมตช์)
* จำนวนคอลัมน์ที่ใช้: 8 คอลัมน์หลัก

1. matches.json -> match_id (Integer): รหัสอ้างอิงการแข่งขัน (ใช้เป็น Key ในการ Merge)
2. matches.json -> home_score, away_score (Integer): ผลการแข่งขันเพื่อคำนวณอัตราการชนะ (Win Rate) รายทีม
3. events.json -> match_id (Integer): รหัสอ้างอิงการแข่งขัน (ใช้เป็น Key ในการ Merge)
4. events.json -> team.name (String): ชื่อทีมที่ทำกิจกรรม (ใช้กรองกลุ่ม High Win-Rate Teams)
5. events.json -> type.name (String): ประเภทเหตุการณ์ คัดเฉพาะเหตุการณ์เชิงพื้นที่ (Pass, Carry)
6. events.json -> location (Array [x, y]): พิกัดจุดเริ่มต้นบนสนาม 120 x 80 หลา (start_x, start_y)
7. events.json -> pass.end_location / carry.end_location (Array [x, y]): พิกัดจุดสิ้นสุดบนสนาม 120 x 80 หลา (end_x, end_y)

---

* ข้อมูล 1 แถว แทนอะไร (unit of analysis)

  ข้อมูลดิบ 1 แถว แทน "1 เหตุการณ์การเคลื่อนที่เชิงพื้นที่ (Single Spatial Movement Event)" ของผู้เล่นในสนามในแมตช์นั้นๆ (เช่น การส่งบอล 1 ครั้ง หรือการพาบอล 1 ครั้ง) ที่บันทึกพิกัดจุดเริ่มต้นและจุดสิ้นสุดไว้อย่างชัดเจน

```python
import pandas as pd
import numpy as np
import requests
from tqdm import tqdm
import concurrent.futures

GITHUB_BASE_URL = "https://raw.githubusercontent.com/statsbomb/open-data/master/data"

# ==========================================
# 1. ดึงรายชื่อ Competitions และ Matches ทั้งหมดที่มีบน GitHub
# ==========================================
print("1. กำลังดึงรายชื่อการแข่งขันและแมตช์ทั้งหมดจาก GitHub...")
comp_url = f"{GITHUB_BASE_URL}/competitions.json"
competitions = requests.get(comp_url).json()

all_match_ids = []
for comp in competitions:
    c_id, s_id = comp['competition_id'], comp['season_id']
    matches_url = f"{GITHUB_BASE_URL}/matches/{c_id}/{s_id}.json"
    res = requests.get(matches_url)
    if res.status_code == 200:
        for m in res.json():
            all_match_ids.append(m['match_id'])

# กำจัด match_id ที่ซ้ำกัน
all_match_ids = list(set(all_match_ids))
print(f"พบแมตช์ทั้งหมดบน GitHub: {len(all_match_ids):,} นัด")

# ==========================================
# 2. ฟังก์ชันดึงและแปลงข้อมูล Events ทุกประเภทที่มีพิกัด X, Y
# ==========================================
def fetch_all_events(match_id):
    url = f"{GITHUB_BASE_URL}/events/{match_id}.json"
    res = requests.get(url)
    if res.status_code != 200:
        return None

    events = res.json()
    extracted = []

    # รวมทุกประเภทเหตุการณ์ที่มีการเคลื่อนที่หรือส่งผลเชิงพื้นที่
    target_types = ['Pass', 'Carry', 'Dribble', 'Shot', 'Pressure', 'Interception', 'Ball Recovery', 'Foul Committed', 'Duel']

    for ev in events:
        event_type = ev.get('type', {}).get('name')
        if event_type in target_types:
            team_name = ev.get('team', {}).get('name')
            player_name = ev.get('player', {}).get('name')

            loc = ev.get('location')
            if not loc or len(loc) < 2:
                continue
            start_x, start_y = loc[0], loc[1]

            # หาพิกัดปลายทาง
            end_x, end_y = start_x, start_y
            if event_type == 'Pass' and 'pass' in ev:
                end_loc = ev['pass'].get('end_location')
                if end_loc and len(end_loc) >= 2:
                    end_x, end_y = end_loc[0], end_loc[1]
            elif event_type == 'Carry' and 'carry' in ev:
                end_loc = ev['carry'].get('end_location')
                if end_loc and len(end_loc) >= 2:
                    end_x, end_y = end_loc[0], end_loc[1]
            elif event_type == 'Shot' and 'shot' in ev:
                end_loc = ev['shot'].get('end_location')
                if end_loc and len(end_loc) >= 2:
                    end_x, end_y = end_loc[0], end_loc[1]

            extracted.append({
                'match_id': match_id,
                'team': team_name,
                'player': player_name,
                'type': event_type,
                'start_x': start_x,
                'start_y': start_y,
                'end_x': end_x,
                'end_y': end_y
            })
    return extracted

# ==========================================
# 3. ยิงดึงทุกแมตช์พร้อมกันด้วย Parallel Processing (30 Threads)
# ==========================================
print(f"2. กำลังดาวน์โหลดข้อมูลจากทุกแมตช์ ({len(all_match_ids):,} นัด) เข้า RAM...")
all_data = []

# ใช้ ThreadPoolExecutor ดึงพร้อมกัน 30 แมตช์ในเวลาเดียวกันเพื่อความเร็วสูงสุด
with concurrent.futures.ThreadPoolExecutor(max_workers=30) as executor:
    results = list(tqdm(executor.map(fetch_all_events, all_match_ids), total=len(all_match_ids)))

for res in results:
    if res:
        all_data.extend(res)

# รวมเป็น DataFrame ตัวเดียวใน RAM
df_all_movements = pd.DataFrame(all_data)

print(f"\nดึงข้อมูลทุกแมตช์สำเร็จ! ได้จำนวนแถวรวมทั้งหมดใน RAM: {len(df_all_movements):,} แถว")
print(f"ขนาด RAM ที่ใช้: {round(df_all_movements.memory_usage().sum() / (1024**2), 2)} MB")
print(df_all_movements.head())
```

```python
import pandas as pd
import requests

GITHUB_BASE_URL = (
    "https://raw.githubusercontent.com/statsbomb/open-data/master/data"
)

# ==========================================
# 1. โหลดเฉพาะตาราง Matches มาคำนวณ Win Rate (แยกตารางเดี่ยว)
# ==========================================
comp_url = f"{GITHUB_BASE_URL}/competitions.json"
competitions = requests.get(comp_url).json()

matches_list = []
for comp in competitions:
  c_id, s_id = comp["competition_id"], comp["season_id"]
  m_url = f"{GITHUB_BASE_URL}/matches/{c_id}/{s_id}.json"
  res = requests.get(m_url)
  if res.status_code == 200:
    for m in res.json():
      home_team = m["home_team"]["home_team_name"]
      away_team = m["away_team"]["away_team_name"]
      home_score = m["home_score"]
      away_score = m["away_score"]

      # บันทึกผู้ชนะในแต่ละนัด
      winner = None
      if home_score > away_score:
        winner = home_team
      elif away_score > home_score:
        winner = away_team

      matches_list.append({
          "match_id": m["match_id"],
          "home_team": home_team,
          "away_team": away_team,
          "winner": winner,
      })

df_matches = pd.DataFrame(matches_list).drop_duplicates(subset=["match_id"])
```

```python
# Invalid / impossible values และ outliers

#------------------------------------------------------
# สนามของ StatsBomb ขนาด 120 x 80
dx = df["end_x"] - df["start_x"]; dy = df["end_y"] - df["start_y"]
dist = np.hypot(dx, dy)

checks = {
    "พิกัดเป็น inf":                      int(np.isinf(df[["start_x","start_y","end_x","end_y"]]).sum().sum()),
    "start_x นอก 0-120":                  int(((df.start_x < 0) | (df.start_x > 120)).sum()),
    "start_y นอก 0-80":                   int(((df.start_y < 0) | (df.start_y > 80)).sum()),
    "end_x นอก 0-120":                    int(((df.end_x < 0) | (df.end_x > 120)).sum()),
    "end_y นอก 0-80":                     int(((df.end_y < 0) | (df.end_y > 80)).sum()),
    "ระยะเคลื่อนที่ > 144.2 (เกินเส้นทแยงสนาม)": int((dist > 144.25).sum()),
}
display(pd.Series(checks, name="จำนวนแถว").to_frame())

# แถวที่จุดเริ่ม = จุดสิ้นสุด (ระยะ 0) แยกตามประเภท event
zero = (dist == 0).groupby(df["type"]).agg(["sum", "mean"])
zero.columns = ["จำนวนที่ระยะ=0", "สัดส่วน"]
display(zero.round(3))

# Outliers ด้วย IQR เฉพาะ Pass และ Carry (event ที่มีจุดปลายทางจริง)
mv = dist[df["type"].isin(["Pass", "Carry"])]
q1, q3 = mv.quantile([0.25, 0.75]); upper = q3 + 1.5 * (q3 - q1)
n_out = (mv > upper).sum()
print(f"ระยะ Pass/Carry: Q1={q1:.1f}, Q3={q3:.1f}, ขอบบน IQR={upper:.1f} | outlier {n_out:,} ({n_out/len(mv)*100:.2f}%)")
display(dist.groupby(df["type"]).describe().round(2))
```

## 5. Data Cleaning
อธิบายทุกการตัดสินใจ **พร้อมเหตุผล** และรายงานจำนวนแถวที่เปลี่ยนแปลงในแต่ละขั้น

| ขั้นตอน | สิ่งที่ทำ | เหตุผล | แถวคงเหลือ |
|---|---|---|---|
| 5.1 | เก็บเฉพาะ `Pass` + `Carry` (ตัด Shot, Pressure, Duel, Interception, Ball Recovery, Foul Committed, Dribble) | คำถามวิจัยคือ "รูปแบบการเคลื่อนที่" ต้องมีจุดเริ่ม-จุดสิ้นสุดจริง เหตุการณ์อื่นไม่มีปลายทาง (`end_x/end_y` เป็น NaN 25.4%) ซึ่งเป็น structural missing ไม่ใช่ MCAR จึงไม่ impute เพราะจะสร้างการเคลื่อนที่ปลอม ส่วน Shot ปลายทางคือกรอบประตู ไม่ใช่การเคลื่อนที่ของผู้เล่น | 6,844,503 |

| 5.3 | ลบแถวที่จุดเริ่ม = จุดสิ้นสุด (ระยะ 0) | Carry ระยะ 0 คือไม่ได้เคลื่อนที่ ไม่มีทิศทาง (มุมนิยามไม่ได้) และจะรวมเป็นกลุ่ม "ไม่ขยับ" ที่บิดผล clustering ส่วน Pass ระยะ 0 น่าจะเป็น `end_location` หายแล้วถูกเติมด้วยจุดเริ่มต้นตอนโหลด | *(ดู `cleaning_log_df`)* |
| 5.4 | ลบแถวซ้ำทุกคอลัมน์ (เก็บแถวแรก) | พิกัด 4 ค่า (ละเอียด 0.1 หลา) + ผู้เล่น + แมตช์ เหมือนกันทุกค่า โอกาสบังเอิญต่ำมาก และมีสัดส่วนน้อยมาก (ข้อจำกัด: ตอนโหลดไม่ได้เก็บ event id) | *(ดู `cleaning_log_df`)* |
| 5.5 | ตรวจ outlier ของระยะ (IQR แยกตาม Pass/Carry) แต่ **คงไว้ทั้งหมด** | outlier คือลูกยาว/ลูกโด่ง/การเลี้ยงบอลยาวที่เกิดขึ้นจริง ไม่มีค่าเกินเส้นทแยงสนาม (144.2 หลา) และเป็นลักษณะที่ใช้แยกสไตล์ตรงไปตรงมา ไปจัดการด้วย log-transform / RobustScaler ใน Pipeline แทน | ไม่เปลี่ยน |
| 5.6 | แก้ชื่อทีมใน events ให้ตรงกับ matches (`name_map`) | value conflict ตอน integration: ชื่อทีมใน events กับ matches เขียนไม่ตรงกัน (เช่น Marseille, Caen, Hyderabad, OL Reign, ATK Mohun Bagan) ถ้าไม่แก้จะ merge ผลการแข่งขันไม่ติด | ไม่เปลี่ยน |
| 5.7 | สร้างผล win/draw/loss ต่อ (แมตช์, ทีม) แล้ว merge เข้า events ด้วย `match_id + team` (m:1) | ใช้คำนวณ Win Rate ของทีม ตรวจแล้วจำนวนแถวก่อน/หลัง merge เท่ากัน และไม่มี `result` เป็น NaN (แมตช์ที่เสมอคิดเป็น draw ไม่ใช่ win) | *(ดู `cleaning_log_df`)* |

> รันเซลล์ด้านล่างแล้วนำตัวเลขแถวคงเหลือจาก `cleaning_log_df` มาใส่ในตาราง

```python
# ==========================================================
# 5. Data Cleaning  (ทุกขั้นบันทึกจำนวนแถวลง cleaning_log)
# ==========================================================

# ลบตัวแปรชั่วคราวจาก Section 4 (index อิงกับ df เดิมทั้งก้อน ถ้าเก็บไว้จะสับสนหลังกรองแถว)
for _v in ["dx", "dy", "dist", "mv", "has_end", "checks", "zero", "miss", "tpm", "q1", "q3", "upper"]:
    globals().pop(_v, None)

cleaning_log = []
def log_step(step, action, n_before, n_after):
    cleaning_log.append({"ขั้นตอน": step, "สิ่งที่ทำ": action,
                         "แถวก่อน": n_before, "แถวหลัง": n_after,
                         "เปลี่ยนแปลง": n_after - n_before})

COORD = ["start_x", "start_y", "end_x", "end_y"]
n0 = len(df_all_movements)            # 9,306,883 (df กับ df_all_movements คือตัวเดียวกัน ไม่ใช่สำเนา)
print(f"เริ่มต้น: {n0:,} แถว")

# ----------------------------------------------------------
# 5.1 เก็บเฉพาะ Pass และ Carry
# ----------------------------------------------------------
# เหตุผล: คำถามคือ "รูปแบบการเคลื่อนที่" ต้องมีจุดเริ่ม-จุดสิ้นสุดจริง
#  - Pressure/Duel/Interception/Ball Recovery/Foul/Dribble ไม่มีจุดปลายทาง (end_x/end_y เป็น NaN 25.4%)
#    เป็น structural missing (ขาดเพราะ "ชนิดเหตุการณ์ไม่มีปลายทาง") ไม่ใช่ MCAR จึงไม่ควร impute
#    เพราะจะสร้างการเคลื่อนที่ปลอมขึ้นมา
#  - Shot มี end_location แต่เป็นจุดที่ลูกยิงไปถึงกรอบประตู ไม่ใช่การเคลื่อนที่ของผู้เล่น
df_c = df_all_movements[df_all_movements["type"].isin(["Pass", "Carry"])].copy()
assert df_c[["end_x", "end_y"]].notna().all().all(), "Pass/Carry ต้องมีจุดปลายทางครบ"
log_step("5.1", "เก็บเฉพาะ Pass + Carry", n0, len(df_c))

# ลดหน่วยความจำ: string -> category, พิกัด -> float32 (ความละเอียด 0.1 หลา float32 เพียงพอ)
for c in ["team", "player", "type"]:
    df_c[c] = df_c[c].astype("category")
df_c[COORD] = df_c[COORD].astype("float32")
df_c = df_c.reset_index(drop=True)
print(f"หลัง 5.1: {len(df_c):,} แถว | หน่วยความจำ {df_c.memory_usage(deep=True).sum()/1e6:,.0f} MB "
      f"(เดิม {df_all_movements.memory_usage(deep=True).sum()/1e6:,.0f} MB)")

# ----------------------------------------------------------
# 5.2 พิกัดนอกสนาม (สนาม StatsBomb = 120 x 80)
# ----------------------------------------------------------
# ตรวจก่อนว่าเกินไปเท่าไร: เกินเล็กน้อย = ความคลาดเคลื่อน/บอลออกนอกสนามจริง -> clip
#                         เกินมาก = ผิดปกติ -> ลบ
TOL = 2.0   # หลา (ปรับได้หลังดูการกระจายของส่วนที่เกินด้านล่าง)
def overshoot(s, hi):
    return np.maximum(-s.to_numpy(), s.to_numpy() - hi).clip(min=0)

over = np.maximum.reduce([overshoot(df_c.start_x, 120), overshoot(df_c.end_x, 120),
                          overshoot(df_c.start_y, 80),  overshoot(df_c.end_y, 80)])
print(f"\nแถวที่มีพิกัดนอกสนาม: {(over > 0).sum():,} ({(over > 0).mean()*100:.4f}%)")
if (over > 0).any():
    print("ระยะที่เกินสนาม (หลา):"); print(pd.Series(over[over > 0]).describe().round(2).to_string())
n_far = int((over > TOL).sum()); n_near = int(((over > 0) & (over <= TOL)).sum())
print(f"เกิน <= {TOL} หลา (clip): {n_near:,} แถว | เกิน > {TOL} หลา (ลบ): {n_far:,} แถว")

n_b = len(df_c)
df_c = df_c[over <= TOL].copy()
for c in ["start_x", "end_x"]:
    df_c[c] = df_c[c].clip(0, 120)
for c in ["start_y", "end_y"]:
    df_c[c] = df_c[c].clip(0, 80)
df_c = df_c.reset_index(drop=True)
log_step("5.2", f"ลบแถวพิกัดเกินสนาม > {TOL} หลา และ clip ที่เหลือให้อยู่ใน 0-120 x 0-80", n_b, len(df_c))

# ----------------------------------------------------------
# 5.3 ระยะเคลื่อนที่ = 0
# ----------------------------------------------------------
# Carry ระยะ 0 = ผู้เล่นไม่ได้เคลื่อนที่ (ไม่มีทิศทาง มุมจะนิยามไม่ได้ และจะรวมเป็นกลุ่ม "ไม่ขยับ" ที่บิดผล clustering)
# Pass ระยะ 0 = น่าจะเป็น end_location หาย แล้ว loader เติมด้วยจุดเริ่มต้น (error จากการโหลด)
dist = np.hypot(df_c.end_x - df_c.start_x, df_c.end_y - df_c.start_y)
print("\nสัดส่วนระยะ = 0 แยกตามชนิด:")
print((dist == 0).groupby(df_c["type"], observed=True).agg(["sum", "mean"]).round(4).to_string())
n_b = len(df_c)
df_c = df_c[dist > 0].reset_index(drop=True)
log_step("5.3", "ลบแถวที่จุดเริ่ม = จุดสิ้นสุด (ระยะ 0)", n_b, len(df_c))

# ----------------------------------------------------------
# 5.4 แถวซ้ำ
# ----------------------------------------------------------
# หมายเหตุ: ตอนโหลดไม่ได้เก็บ event id / เวลา จึงแยก "ซ้ำจริง" กับ "บังเอิญเหมือนกัน" ไม่ได้ 100%
# แต่พิกัด 4 ค่าละเอียด 0.1 หลา + ผู้เล่น + แมตช์ เหมือนกันทุกค่า เป็นไปได้น้อยมากที่จะบังเอิญ
n_b = len(df_c)
n_dup = int(df_c.duplicated().sum())
print(f"\nแถวซ้ำทุกคอลัมน์: {n_dup:,} ({n_dup/n_b*100:.3f}%)")
df_c = df_c.drop_duplicates().reset_index(drop=True)
log_step("5.4", "ลบแถวซ้ำทุกคอลัมน์ (เก็บแถวแรก)", n_b, len(df_c))

# ----------------------------------------------------------
# 5.5 Outlier ของระยะเคลื่อนที่ (IQR แยกตามชนิด) -> รายงานแต่ "ไม่ลบ"
# ----------------------------------------------------------
dist = np.hypot(df_c.end_x - df_c.start_x, df_c.end_y - df_c.start_y)
g = dist.groupby(df_c["type"], observed=True)
q = g.quantile([0.25, 0.75]).unstack()
upper_bound = q[0.75] + 1.5 * (q[0.75] - q[0.25])
is_out = dist > df_c["type"].astype(str).map(upper_bound).astype(float)
rep = pd.DataFrame({"Q1": q[0.25], "Q3": q[0.75], "ขอบบน IQR": upper_bound,
                    "จำนวน outlier": is_out.groupby(df_c["type"], observed=True).sum(),
                    "สัดส่วน": is_out.groupby(df_c["type"], observed=True).mean()}).round(3)
print("\nOutlier ของระยะเคลื่อนที่ (IQR แยกตามชนิด):"); print(rep.to_string())
print(f"ระยะสูงสุด {dist.max():.1f} หลา (เส้นทแยงสนาม = 144.2) -> ไม่มีค่าเป็นไปไม่ได้")
# ตัดสินใจ: ไม่ลบ เพราะเป็นเหตุการณ์จริง (ลูกโด่ง/ลูกยาวจากผู้รักษาประตู/ตั้งเตะ/เลี้ยงบอลยาว)
# และเป็นลักษณะที่ใช้แยก archetype การเล่นแบบตรงไปตรงมา ไปจัดการด้วย log-transform/RobustScaler ใน Pipeline
log_step("5.5", "ตรวจ outlier ระยะ (IQR) แต่คงไว้ทั้งหมด", len(df_c), len(df_c))

# ----------------------------------------------------------
# 5.6 ชื่อทีมไม่ตรงกันระหว่าง events กับ matches (value conflict ตอนรวมข้อมูล)
# ----------------------------------------------------------
# พบว่า mapping ชื่อทีมแบบ "ทั้งชุด" ใช้ไม่ได้ เพราะชื่อซ้ำข้ามทีมชาย/หญิง
# (เช่น events ใช้ 'Manchester United' ทั้งชายและหญิง แต่ matches ใช้ 'Manchester United W' เฉพาะทีมหญิง)
# จึงจับคู่ "ทีละแมตช์": ทีมใน events -> ชื่อใน matches (canon) แล้วใช้ชื่อจาก matches เป็นชื่อมาตรฐาน
ev_pairs = df_c[["match_id", "team"]].drop_duplicates()
ev_pairs["team"] = ev_pairs["team"].astype(str)
pairs = ev_pairs.merge(df_matches[["match_id", "home_team", "away_team", "winner"]],
                       on="match_id", how="left", validate="m:1").reset_index(drop=True)
pairs["canon"] = np.where((pairs.team == pairs.home_team) | (pairs.team == pairs.away_team), pairs.team, None)
n_direct = int(pairs.canon.notna().sum())
print(f"\nคู่ (แมตช์, ทีม) ทั้งหมด {len(pairs):,} | ชื่อตรงกับ matches ทันที {n_direct:,} | ไม่ตรง {len(pairs)-n_direct:,}")

hints = {}   # ชื่อใน events -> ชื่อใน matches ที่เคยเจอจากแมตช์อื่น (ใช้เฉพาะเมื่อชื่อนั้นอยู่ในแมตช์นี้จริง)
for _round in range(10):
    changed = False
    todo = pairs.loc[pairs.canon.isna(), "match_id"].unique()
    for mid, gg in pairs[pairs.match_id.isin(todo)].groupby("match_id"):
        home, away = gg.home_team.iloc[0], gg.away_team.iloc[0]
        for i in gg.index[gg.canon.isna()]:                      # 1) ลองใช้ hint จากแมตช์อื่น
            cands = hints.get(pairs.at[i, "team"], set()) & {home, away}
            if len(cands) == 1:
                pairs.at[i, "canon"] = next(iter(cands)); changed = True
        gg = pairs.loc[gg.index]
        unres, res = gg[gg.canon.isna()], gg[gg.canon.notna()]
        left = [n for n in (home, away) if n not in set(res.canon)]
        if len(unres) == 1 and len(left) == 1:                   # 2) ทีมที่เหลือ = ชื่อที่เหลือใน matches
            pairs.at[unres.index[0], "canon"] = left[0]; changed = True
    for t, c in pairs.loc[pairs.canon.notna() & (pairs.team != pairs.canon), ["team", "canon"]].itertuples(index=False):
        hints.setdefault(t, set()).add(c)
    if not changed:
        break

diff = pairs[pairs.canon.notna() & (pairs.team != pairs.canon)]
print("ชื่อที่ถูกแก้ (events -> matches) และจำนวนแมตช์:")
print(diff.groupby(["team", "canon"]).size().rename("แมตช์").to_string())
unresolved = pairs[pairs.canon.isna()]
print(f"ยังจับคู่ไม่ได้: {len(unresolved)} คู่")
if len(unresolved):
    print(unresolved[["match_id", "team", "home_team", "away_team"]].to_string())   # ตรวจด้วยมือแล้วเติมใน pairs.canon ได้

# ----------------------------------------------------------
# 5.7 ผลการแข่งขันต่อทีมต่อแมตช์ แล้ว merge เข้า events (key = match_id + team ใน events)
# ----------------------------------------------------------
# winner เป็น None = เสมอ -> draw
pairs["result"] = np.where(pairs.canon.isna(), None,
                    np.where(pairs.winner.isna(), "draw", np.where(pairs.winner == pairs.canon, "win", "loss")))
print("\nสัดส่วนผลการแข่งขัน (ระดับทีม-แมตช์):")
print(pairs["result"].value_counts(normalize=True).round(3).to_string())

key = pairs[["match_id", "team", "canon", "result"]].copy()
assert not key.duplicated(["match_id", "team"]).any()
key["team"] = pd.Categorical(key["team"], categories=df_c["team"].cat.categories)   # key ชนิดเดียวกัน merge เร็ว ประหยัด RAM

n_b = len(df_c)
df_clean = df_c.merge(key, on=["match_id", "team"], how="left", validate="m:1")
assert len(df_clean) == n_b
print(f"\nแถวก่อน merge: {n_b:,} | หลัง merge: {len(df_clean):,} | จับคู่ชื่อทีมไม่ได้: {df_clean['canon'].isna().sum():,}")
log_step("5.6-5.7", "จับคู่ชื่อทีมเป็นรายแมตช์ + merge ผลแข่งด้วย match_id + team (m:1)", n_b, len(df_clean))

# ใช้ชื่อจาก matches เป็นชื่อทีมมาตรฐาน (แยกทีมชาย/หญิงได้) และตัดแมตช์ที่จับคู่ไม่ได้ (ถ้ามี)
n_b = len(df_clean)
df_clean = df_clean[df_clean["canon"].notna()].copy()
df_clean["team"] = df_clean["canon"].astype("category")
df_clean["result"] = df_clean["result"].astype("category")
df_clean = df_clean.drop(columns="canon").reset_index(drop=True)
log_step("5.7", "ใช้ชื่อทีมจาก matches เป็นมาตรฐาน และตัดแถวที่จับคู่ชื่อทีมไม่ได้", n_b, len(df_clean))
assert df_clean["result"].notna().all()

# ----------------------------------------------------------
# สรุปผล
# ----------------------------------------------------------
cleaning_log_df = pd.DataFrame(cleaning_log)
display(cleaning_log_df)
print(f"\nข้อมูลสะอาด df_clean: {df_clean.shape[0]:,} แถว x {df_clean.shape[1]} คอลัมน์ "
      f"(เหลือ {len(df_clean)/n0*100:.1f}% ของข้อมูลดิบ) | {df_clean.memory_usage(deep=True).sum()/1e6:,.0f} MB")
print(df_clean.dtypes.to_string())
df_clean.head()
```

```python
# Cell 0: Setup EDA
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

# เลือกข้อมูลจาก eda ที่มีอยู่แล้ว หรือสร้างจาก df_clean
if "eda" in globals():
    eda = eda.copy()
elif "df_clean" in globals():
    eda = df_clean.copy()
else:
    raise NameError(
        "ไม่พบ eda หรือ df_clean กรุณารันขั้นตอน Cleaning ก่อน"
    )

# ตรวจสอบคอลัมน์ที่จำเป็น
required_cols = [
    "match_id", "team", "type", "result",
    "start_x", "start_y", "end_x", "end_y"
]

missing_cols = [c for c in required_cols if c not in eda.columns]
if missing_cols:
    raise ValueError(f"ไม่พบคอลัมน์ที่จำเป็น: {missing_cols}")

# สร้างตัวแปรการเคลื่อนที่ หากยังไม่มี
eda["dx"] = eda["end_x"] - eda["start_x"]
eda["dy"] = eda["end_y"] - eda["start_y"]
eda["dist"] = np.hypot(eda["dx"], eda["dy"])
eda["angle"] = np.degrees(np.arctan2(eda["dy"], eda["dx"]))

# สร้างทิศทางการเคลื่อนที่ตามแกน X
# Forward = เคลื่อนที่ไปทางแกน X บวก
# Lateral = เคลื่อนที่ในแนวแกน Y เป็นหลัก
# Backward = เคลื่อนที่ไปทางแกน X ลบ

abs_angle = eda["angle"].abs()

eda["direction"] = pd.cut(
    abs_angle,
    bins=[-0.1, 45, 135, 180.1],
    labels=["Forward", "Lateral", "Backward"]
)

# แบ่งพื้นที่สนามตามแกน X
eda["start_third"] = pd.cut(
    eda["start_x"],
    bins=[-0.1, 40, 80, 120.1],
    labels=["Defensive", "Middle", "Final"]
)

eda["end_third"] = pd.cut(
    eda["end_x"],
    bins=[-0.1, 40, 80, 120.1],
    labels=["Defensive", "Middle", "Final"]
)

sns.set_theme(style="whitegrid", context="notebook")

print(f"จำนวน Events: {len(eda):,}")
print(f"จำนวน Matches: {eda['match_id'].nunique():,}")
print(f"จำนวน Teams: {eda['team'].nunique():,}")
print("เตรียมข้อมูล EDA เรียบร้อย")
```

**โน้ต**

**Pass (การส่งบอล)**
* ความหมาย: คือการเตะหรือส่งลูกฟุตบอลจากผู้เล่นคนหนึ่งไปยังเพื่อนร่วมทีม เพื่อรักษาการครองบอล เปลี่ยนพื้นที่เล่น หรือเจาะแนวรับ
* ลักษณะการใช้งาน: มีหลายรูปแบบ เช่น การส่งสั้น (Short pass), การวางบอลยาว (Long pass), การส่งแบบสัมผัสเดียว (One-touch pass), หรือการจ่ายบอลทะลุช่องเพื่อทำลายแนวรับ (Killer Pass)
* เป้าหมาย: ใช้ความเร็วของลูกบอลที่เคลื่อนที่ผ่านอากาศหรือเรียดไปกับพื้น เพื่อส่งต่อความได้เปรียบให้เพื่อนเล่นต่อได้ทันที


**Carry (การพาบอล)**
* ความหมาย: คือการที่ผู้เล่นครอบครองลูกบอลแล้ว "พาบอลด้วยเท้าตัวเอง" วิ่งเคลื่อนที่ขึ้นไปข้างหน้าผ่านพื้นที่ว่าง
* ข้อแตกต่างจาก Dribble (การเลี้ยงหลบ):

	• Dribble เน้นการเลี้ยงหลบ 1-ต่อ-1 เพื่อผ่านตัวประกบหรือคู่ต่อสู้โดยตรง

	• Carry เน้นการ "ลำเลียงบอล" หรือพาบอลรุกคืบเข้าไปในแดนคู่ต่อสู้เพื่อกินพื้นที่ (ไม่จำเป็นต้องมีการเผชิญหน้าหรือหลบผู้เล่นฝ่ายตรงข้ามเสมอไป)
* Progressive Carry (การพาบอลก้าวหน้า): ในสถิติฟุตบอลยุคใหม่ หมายถึงการพาบอลขึ้นหน้าอย่างมีความหมาย (เช่น ระยะทางอย่างน้อย 10 หลาในแดนทั่วไป หรือ 5 หลาเมื่อเข้าใกล้กรอบเขตโทษ) เพื่อสร้างจังหวะรุกให้ทีม



```python
# EDA: Distribution of clustering features
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
plot_parts = []

for event_type in ["Pass", "Carry"]:
    s = eda.loc[
        eda["type"].astype(str).eq(event_type),
        ["dist", "angle"]
    ].dropna()

    # สุ่มเฉพาะสำหรับการวาดกราฟ ไม่ได้แก้ข้อมูล eda
    s = s.sample(n=min(150_000, len(s)), random_state=42).copy()
    s["type"] = event_type
    plot_parts.append(s)

plot_sample = __import__("pandas").concat(
    plot_parts, ignore_index=True
)

fig, ax = plt.subplots(1, 2, figsize=(13, 4))

for event_type in ["Pass", "Carry"]:
    s = plot_sample[plot_sample["type"] == event_type]

    ax[0].hist(
        np.log1p(s["dist"]),
        bins=60, alpha=0.55, density=True,
        label=event_type
    )
    ax[1].hist(
        s["angle"],
        bins=np.linspace(-180, 180, 73),
        alpha=0.55, density=True,
        label=event_type
    )

ax[0].set_title("Distribution of log1p(Distance)")
ax[0].set_xlabel("log1p(Distance)")
ax[0].set_ylabel("Density")
ax[0].legend()

ax[1].set_title("Distribution of Movement Angle")
ax[1].set_xlabel("Angle (degrees)")
ax[1].set_ylabel("Density")
ax[1].set_xlim(-180, 180)
ax[1].legend()

plt.tight_layout()
plt.show()
```

**สรุปผลการวิเคราะห์การกระจายตัวของ Distance และ Movement Angle**

จากกราฟการกระจายตัวของระยะทาง พบว่า Pass มีค่า log1p(Distance) โดยรวมสูงกว่า Carry สะท้อนว่า Pass มีระยะทางต่อเหตุการณ์มากกว่า Carry โดยทั่วไป ขณะที่ Carry มีการกระจายตัวไปทางระยะทางที่สั้นกว่า การแปลงระยะทางด้วย log1p() ช่วยลดความเบ้ของข้อมูลและทำให้เห็นการกระจายตัวได้ชัดเจนขึ้น

สำหรับกราฟ Movement Angle พบว่าทั้ง Pass และ Carry มีการกระจายตัวของมุมในหลายทิศทาง ซึ่งช่วยอธิบายรูปแบบการเปลี่ยนตำแหน่งของแต่ละเหตุการณ์

**สรุป:** ระยะทางและมุมเป็นคุณลักษณะที่ช่วยอธิบายรูปแบบการเคลื่อนที่ได้แตกต่างกัน จึงนำมาพิจารณาในขั้นตอน Feature Engineering โดยแปลงระยะทางด้วย log1p() และแปลงมุมเป็นค่า sine และ cosine เพื่อใช้เป็น Feature สำหรับ Clustering

```python
#  Movement patterns by team-match result

# ใช้เฉพาะ Pass และ Carry
tc = eda[eda["type"].isin(["Pass", "Carry"])].copy()

# ทำให้ชื่อผลการแข่งขันเป็นตัวพิมพ์เล็กเพื่อจัดกลุ่มให้สม่ำเสมอ
tc["_result"] = tc["result"].astype("string").str.strip().str.lower()

# ตรวจสอบว่ามีผลการแข่งขันที่ใช้งานได้หรือไม่
print("ค่าที่พบใน result:")
display(tc["_result"].value_counts(dropna=False).to_frame("Count"))

# คำนวณคุณลักษณะของแต่ละทีมในแต่ละแมตช์
team_match = (
    tc.groupby(["match_id", "team", "_result"], observed=True)
    .agg(
        Total_Events=("type", "size"),
        Pass_Count=("type", lambda s: (s == "Pass").sum()),
        Carry_Count=("type", lambda s: (s == "Carry").sum()),
        Mean_Pass_Distance=(
            "dist",
            lambda s: np.nan  # คอลัมน์นี้จะคำนวณแยกด้านล่าง
        )
    )
    .reset_index()
)

# คำนวณระยะ Pass และ Carry แยกตามทีม-แมตช์
pass_distance = (
    tc[tc["type"] == "Pass"]
    .groupby(["match_id", "team", "_result"], observed=True)["dist"]
    .mean()
    .rename("Mean_Pass_Distance")
    .reset_index()
)

carry_distance = (
    tc[tc["type"] == "Carry"]
    .groupby(["match_id", "team", "_result"], observed=True)["dist"]
    .mean()
    .rename("Mean_Carry_Distance")
    .reset_index()
)

forward_pct = (
    tc.assign(_is_forward=tc["direction"].astype("string").str.lower().eq("forward"))
    .groupby(["match_id", "team", "_result"], observed=True)["_is_forward"]
    .mean()
    .mul(100)
    .rename("Forward_Movement_Pct")
    .reset_index()
)

team_match = team_match.drop(columns=["Mean_Pass_Distance"]).merge(
    pass_distance, on=["match_id", "team", "_result"], how="left"
).merge(
    carry_distance, on=["match_id", "team", "_result"], how="left"
).merge(
    forward_pct, on=["match_id", "team", "_result"], how="left"
)

# แปลงจำนวน Pass/Carry เป็นเปอร์เซ็นต์ของ Events ในทีม-แมตช์นั้น
team_match["Pass_Pct"] = (
    team_match["Pass_Count"] / team_match["Total_Events"] * 100
)
team_match["Carry_Pct"] = (
    team_match["Carry_Count"] / team_match["Total_Events"] * 100
)

# ยอมรับเฉพาะผลการแข่งขันที่รู้จัก
valid_results = ["win", "draw", "loss"]
team_match_valid = team_match[
    team_match["_result"].isin(valid_results)
].copy()

if team_match_valid.empty:
    print("ไม่พบค่า result เป็น win/draw/loss หลังแปลงตัวพิมพ์เล็ก")
else:
    result_table = (
        team_match_valid.groupby("_result", observed=True)
        .agg(
            Team_Match_Count=("match_id", "size"),
            Mean_Pass_Pct=("Pass_Pct", "mean"),
            Mean_Carry_Pct=("Carry_Pct", "mean"),
            Mean_Pass_Distance=("Mean_Pass_Distance", "mean"),
            Mean_Carry_Distance=("Mean_Carry_Distance", "mean"),
            Mean_Forward_Pct=("Forward_Movement_Pct", "mean")
        )
        .reindex(["win", "draw", "loss"])
        .rename_axis("Result")
        .reset_index()
    )

    result_table = result_table.rename(columns={
        "Team_Match_Count": "จำนวนทีม-แมตช์",
        "Mean_Pass_Pct": "Pass เฉลี่ย (%)",
        "Mean_Carry_Pct": "Carry เฉลี่ย (%)",
        "Mean_Pass_Distance": "ระยะ Pass เฉลี่ย",
        "Mean_Carry_Distance": "ระยะ Carry เฉลี่ย",
        "Mean_Forward_Pct": "Forward เฉลี่ย (%)"
    })

    numeric_cols = result_table.select_dtypes(include="number").columns
    result_table[numeric_cols] = result_table[numeric_cols].round(2)

    display(result_table)
```

สรุปผลการวิเคราะห์รูปแบบการเคลื่อนที่ตามผลการแข่งขัน

จากการเปรียบเทียบข้อมูลระหว่างทีมที่ชนะ เสมอ และแพ้ พบว่าทีมที่ชนะมีสัดส่วน Pass เฉลี่ย 59.05% และ Carry เฉลี่ย 40.95% ส่วนทีมที่เสมอและแพ้มีสัดส่วน Pass ใกล้เคียงกันที่ 59.68% และ 59.75% ตามลำดับ

ด้านระยะทาง Pass เฉลี่ย ทีมที่ชนะมีค่า 21.14 หน่วย ขณะที่ทีมที่เสมอและแพ้มีค่า 21.94 และ 21.93 หน่วยตามลำดับ ส่วนระยะทาง Carry เฉลี่ยมีค่าใกล้เคียงกันทุกกลุ่ม

สำหรับการเคลื่อนที่ไปข้างหน้า (Forward) ทีมที่ชนะมีค่าเฉลี่ย 37.54% ต่ำกว่าทีมที่เสมอ (39.10%) และทีมที่แพ้ (39.25%) เล็กน้อย

**สรุป:** รูปแบบการเคลื่อนที่มีความแตกต่างกันระหว่างกลุ่มผลการแข่งขัน แต่ข้อมูลนี้เพียงอย่างเดียวยังไม่สามารถสรุปได้ว่ารูปแบบใดเป็นสาเหตุให้ทีมชนะหรือแพ้

```python
# EDA: Compare Forward Movement between Win and Loss
from scipy import stats
import numpy as np
import pandas as pd

# เลือกข้อมูลระดับทีม-แมตช์
w = (
    team_match_valid.loc[
        team_match_valid["_result"].eq("win"),
        "Forward_Movement_Pct"
    ].dropna()
)

l = (
    team_match_valid.loc[
        team_match_valid["_result"].eq("loss"),
        "Forward_Movement_Pct"
    ].dropna()
)

if len(w) < 2 or len(l) < 2:
    print("ข้อมูลไม่เพียงพอสำหรับการทดสอบ")
else:
    # Welch's t-test
    test = stats.ttest_ind(w, l, equal_var=False)

    # Cohen's d: pooled standard deviation
    pooled_sd = np.sqrt(
        ((len(w) - 1) * w.var(ddof=1)
         + (len(l) - 1) * l.var(ddof=1))
        / (len(w) + len(l) - 2)
    )

    d = (w.mean() - l.mean()) / pooled_sd if pooled_sd > 0 else np.nan

    result_test = pd.DataFrame({
        "Metric": [
            "จำนวนทีม-แมตช์",
            "Forward Movement เฉลี่ย (%)",
            "Standard Deviation",
            "Welch t-test p-value",
            "Cohen's d"
        ],
        "Win": [
            len(w), w.mean(), w.std(ddof=1), np.nan, np.nan
        ],
        "Loss": [
            len(l), l.mean(), l.std(ddof=1), np.nan, np.nan
        ]
    })

    display(result_test.round(4))

    print(f"ผลต่างค่าเฉลี่ย Win - Loss: {w.mean() - l.mean():.3f} จุดเปอร์เซ็นต์")
    print(f"Welch's t-test p-value: {test.pvalue:.4e}")
    print(f"Cohen's d: {d:.3f}")
```


จากการเปรียบเทียบค่า Forward Movement เฉลี่ย พบว่าทีมชนะมีค่าเฉลี่ย 37.539% ขณะที่ทีมแพ้มีค่าเฉลี่ย 39.248% โดยทีมชนะมีค่าน้อยกว่าประมาณ 1.709 จุดเปอร์เซ็นต์

ผลการทดสอบ Welch’s t-test ได้ค่า p-value = 3.8315 × 10⁻⁴⁴ ซึ่งน้อยกว่า 0.05 แสดงว่าค่าเฉลี่ย Forward Movement ของทั้งสองกลุ่มแตกต่างกันอย่างมีนัยสำคัญทางสถิติ นอกจากนี้ Cohen’s d = -0.358 บ่งชี้ว่าความแตกต่างมีขนาดเล็กถึงปานกลาง โดยทีมชนะมีค่าเฉลี่ยต่ำกว่าทีมแพ้

**สรุป:** ผลการทดสอบพบความแตกต่างของ Forward Movement ระหว่างทีมชนะและทีมแพ้อย่างมีนัยสำคัญทางสถิติ อย่างไรก็ตาม ผลดังกล่าวไม่ได้ยืนยันว่าการเคลื่อนที่ไปข้างหน้าน้อยกว่าจะทำให้ชนะ เพราะยังมีปัจจัยอื่นที่เกี่ยวข้องกับผลการแข่งขัน

```python

# Pitch heatmaps for Pass/Carry start and end locations
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle, Arc

# 1. ฟังก์ชันวาดเส้นสนามฟุตบอล
def draw_pitch(ax):
    # ขอบสนาม
    line_color = "#BDBDBD"
    lw = 2

    ax.add_patch(Rectangle(
        (0, 0), 120, 80,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))

    # เส้นแบ่งครึ่งสนาม
    ax.plot([60, 60], [0, 80],
            color=line_color, linewidth=lw, zorder=3)

    # วงกลมกลางสนาม
    ax.add_patch(Circle(
        (60, 40), 10,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))

    # จุดกลางสนาม
    ax.scatter([60], [40], color=line_color, s=8, zorder=4)

    # เขตโทษทั้งสองฝั่ง
    ax.add_patch(Rectangle(
        (0, 18), 18, 44,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))
    ax.add_patch(Rectangle(
        (102, 18), 18, 44,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))

    # เขตประตูทั้งสองฝั่ง
    ax.add_patch(Rectangle(
        (0, 30), 6, 20,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))
    ax.add_patch(Rectangle(
        (114, 30), 6, 20,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))

    # จุดยิงจุดโทษ
    ax.scatter([12, 108], [40, 40],
               color=line_color, s=8, zorder=4)

    # ส่วนโค้งหน้ากรอบเขตโทษ
    ax.add_patch(Arc(
        (12, 40), 20, 20,
        theta1=307, theta2=53,
        color=line_color, linewidth=lw, zorder=3
    ))
    ax.add_patch(Arc(
        (108, 40), 20, 20,
        theta1=127, theta2=233,
        color=line_color, linewidth=lw, zorder=3
    ))

    # กำหนดขอบเขตและสัดส่วนสนาม
    ax.set_xlim(0, 120)
    ax.set_ylim(0, 80)
    ax.set_aspect("equal", adjustable="box")


# 2. เตรียมข้อมูลสำหรับวาด Heatmap
plot_df = eda[
    eda["type"].isin(["Pass", "Carry"])
    & eda[["start_x", "start_y", "end_x", "end_y"]]
        .notna().all(axis=1)
].copy()

plot_df = plot_df.sample(
    n=min(200_000, len(plot_df)),
    random_state=425
)

# 3. วาด Heatmap ทั้ง 4 รูป
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

plot_specs = [
    ("Pass", "start_x", "start_y", axes[0, 0],
     "Pass - Start Location"),
    ("Pass", "end_x", "end_y", axes[0, 1],
     "Pass - End Location"),
    ("Carry", "start_x", "start_y", axes[1, 0],
     "Carry - Start Location"),
    ("Carry", "end_x", "end_y", axes[1, 1],
     "Carry - End Location")
]

for event_type, x_col, y_col, ax, title in plot_specs:
    subset = plot_df[plot_df["type"] == event_type]

    if subset.empty:
        ax.set_title(f"{title} (No data)")
        continue

    # Heatmap
    hist = ax.hist2d(
        subset[x_col],
        subset[y_col],
        bins=[24, 16],
        range=[[0, 120], [0, 80]],
        cmap="YlOrRd",
        cmin=1,
        zorder=1
    )

    # วาดเส้นสนามทับลงบน Heatmap
    draw_pitch(ax)

    ax.set_xlabel("Pitch X")
    ax.set_ylabel("Pitch Y")
    ax.set_title(title)

    fig.colorbar(
        hist[3],
        ax=ax,
        label="Number of Events",
        fraction=0.046,
        pad=0.03
    )

plt.suptitle(
    "Spatial Distribution of Pass and Carry",
    fontsize=15
)
plt.tight_layout()
plt.show()
```

สรุปผลการวิเคราะห์ Heatmap ตำแหน่ง Pass และ Carry

**1. Pass – Start Location (ตำแหน่งเริ่มส่งบอล)**  
พบการส่งบอลหนาแน่นบริเวณใกล้กรอบเขตโทษฝั่งซ้ายของภาพ และบริเวณกลางสนาม แสดงว่าจุดเริ่มส่งบอลที่พบได้บ่อยอยู่ในพื้นที่เหล่านี้

**2. ตำแหน่งและทิศทางการเคลื่อนที่**

ทิศทาง Lateral มีสัดส่วนสูงสุด 47.90% รองลงมาคือ Forward 37.45% และ Backward 14.66% ขณะที่ Transition Matrix แสดงว่า Pass จาก Defensive Third ไป Middle Third มีสัดส่วน 38.2% และจาก Middle Third ไป Final Third มีสัดส่วน 20.7% ส่วน Carry จาก Middle Third ยังคงอยู่ในโซนเดิม 89.3%

ตัวเลขใน Transition Matrix เป็นร้อยละของเหตุการณ์ที่เริ่มจากโซนนั้น ๆ จึงใช้เปรียบเทียบแนวโน้มการเปลี่ยนโซนได้ ผลการวิเคราะห์สนับสนุนให้ใช้พิกัดเริ่มต้น–ปลายทาง ระยะทาง และทิศทางเป็น Feature สำหรับ Clustering โดยแปลงมุมเป็น sin(angle) และ cos(angle) เพื่อแทนทิศทางที่ต่อเนื่องกันบริเวณ -180° และ 180°

**3. ความแตกต่างของ Forward Movement ระหว่างทีมชนะและทีมแพ้**

ทีมชนะมี Forward Movement เฉลี่ย 37.539% (SD = 4.957) ส่วนทีมแพ้มีค่าเฉลี่ย 39.248% (SD = 4.586) โดยทีมชนะมีค่าเฉลี่ยต่ำกว่า 1.709 จุดเปอร์เซ็นต์ ผล Welch’s t-test ได้ p-value = 3.8315 × 10⁻⁴⁴ ซึ่งต่ำกว่า 0.05 จึงมีหลักฐานทางสถิติว่าค่าเฉลี่ยของทั้งสองกลุ่มแตกต่างกันในข้อมูลนี้ ส่วน Cohen’s d = -0.358 แสดงว่าค่าเฉลี่ยของทีมชนะต่ำกว่าทีมแพ้ประมาณ 0.358 ส่วนเบี่ยงเบนมาตรฐาน

อย่างไรก็ตาม p-value ไม่ได้บอกว่าความแตกต่างมีความสำคัญเชิงแท็กติกมากเพียงใด และไม่สามารถสรุปเหตุและผลได้ นอกจากนี้ ข้อมูลของทีมที่แข่งขันในแมตช์เดียวกันอาจมีความเกี่ยวข้องกัน จึงควรใช้ผลนี้เป็นข้อมูลประกอบการวิเคราะห์ ไม่ใช่ข้อยืนยันว่าการเคลื่อนที่ไปข้างหน้ามากขึ้นทำให้ทีมแพ้

```python
# Start-third to end-third transition matrices

zones = ["Defensive", "Middle", "Final"]

transition_data = eda[
    eda["type"].isin(["Pass", "Carry"])
    & eda["start_third"].notna()
    & eda["end_third"].notna()
].copy()

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax, event_type in zip(axes, ["Pass", "Carry"]):
    subset = transition_data[transition_data["type"] == event_type]

    if subset.empty:
        ax.set_title(f"{event_type} (No data)")
        continue

    counts = pd.crosstab(
        subset["start_third"],
        subset["end_third"],
        dropna=False
    )

    counts = counts.reindex(index=zones, columns=zones, fill_value=0)

    # คิดเป็นเปอร์เซ็นต์ภายในแต่ละพื้นที่เริ่มต้น
    row_totals = counts.sum(axis=1).replace(0, np.nan)
    row_pct = counts.div(row_totals, axis=0) * 100
    row_pct = row_pct.fillna(0)

    sns.heatmap(
        row_pct,
        annot=True,
        fmt=".1f",
        cmap="YlGnBu",
        vmin=0,
        vmax=100,
        linewidths=0.5,
        cbar_kws={"label": "Percentage within start zone (%)"},
        ax=ax
    )

    ax.set_title(f"{event_type}: Start Third → End Third")
    ax.set_xlabel("End Third")
    ax.set_ylabel("Start Third")

plt.tight_layout()
plt.show()
```

การส่งบอลที่เริ่มจาก Defensive Third ส่วนใหญ่ 57.6% ยังคงอยู่ในโซนเดิม และ 38.2% เคลื่อนที่ไป Middle Third ส่วนการส่งบอลจาก Middle Third พบว่า 71.2% อยู่ในโซนเดิม และ 20.7% เคลื่อนที่ไป Final Third ขณะที่ Pass ที่เริ่มจาก Final Third ส่วนใหญ่ 88.5% ยังคงอยู่ในโซนเดิม

**2. Carry: Start Third → End Third**  
การพาบอลมีแนวโน้มอยู่ในโซนเดิมมากกว่า Pass โดย Carry จาก Defensive Third อยู่ในโซนเดิม 84.6% และจาก Middle Third อยู่ในโซนเดิม 89.3% ส่วน Carry ที่เริ่มจาก Final Third ยังคงอยู่ในโซนเดิมสูงถึง 96.1%

## 7. Feature Engineering
| Feature | สร้างจาก | เหตุผล |
|---|---|---|
| `start_x` | `start_x` (ใช้ตรง ๆ) | แสดงตำแหน่งเริ่มต้นตามแนวยาวของสนาม ช่วยแยกการเคลื่อนที่ในแดนรับ แดนกลาง และแดนบุก |
| `start_y` | `start_y` (ใช้ตรง ๆ) | แสดงตำแหน่งเริ่มต้นตามแนวกว้างของสนาม ช่วยแยกการเล่นบริเวณริมเส้นและบริเวณกลางสนาม |
| `log_dist` | `log1p(hypot(end_x-start_x, end_y-start_y))` | ระยะทางมีการกระจายแบบเบ้ขวา การแปลงด้วย log ช่วยลดอิทธิพลของการเคลื่อนที่ระยะไกลและค่าผิดปกติต่อ K-Means |
| `angle_sin` | `(end_y-start_y) / dist` | ใช้แทนองค์ประกอบแนวตั้งของทิศทางการเคลื่อนที่ โดยแปลงมุมเป็นค่า sine เพื่อให้แทนทิศทางได้เหมาะสม |
| `angle_cos` | `(end_x-start_x) / dist` | ใช้คู่กับ angle_sin เพื่อแทนองค์ประกอบแนวนอนของทิศทางการเคลื่อนที่ ค่าบวกหมายถึงเคลื่อนที่ไปทางแกน X บวก และค่าลบหมายถึงเคลื่อนที่ไปทางแกน X ลบ |
| `is_pass` | `type == "Pass"` (Pass = 1, Carry = 0) | ช่วยให้โมเดลแยกประเภทการเคลื่อนที่ระหว่างการส่งบอลกับการพาบอล ซึ่งอาจมีรูปแบบและระยะทางแตกต่างกัน |
| `goal_progress` | ระยะจากจุดเริ่มต้นถึงประตู `(120, 40)` − ระยะจากจุดสิ้นสุดถึงประตู `(120, 40)` | วัดการเปลี่ยนแปลงของระยะห่างจากจุดอ้างอิง (120, 40) ค่าบวกหมายถึงจุดสิ้นสุดอยู่ใกล้จุดอ้างอิงกว่าจุดเริ่มต้น ส่วนค่าลบหมายถึงจุดสิ้นสุดอยู่ไกลขึ้น ทั้งนี้ การตีความว่าเข้าใกล้ประตูคู่แข่งต้องตรวจสอบระบบพิกัดของข้อมูลต้นทาง|

**ตัวแปรที่ไม่นำมาใช้เป็น Feature**

- `end_x`, `end_y`, `dx`, `dy` — ไม่นำมาใช้โดยตรง เพราะใช้คำนวณ Feature ด้านระยะทางและทิศทางแล้ว
- `start_third`, `end_third`, `direction` — เป็นตัวแปรที่จัดกลุ่มหรือสรุปจากตำแหน่งและทิศทาง จึงเก็บไว้ช่วยตีความผล
- `result`, `team`, `player`, `match_id` — เก็บไว้ใช้วิเคราะห์และตีความผลการจัดกลุ่ม ไม่ใช้เป็น Input ของโมเดล

**การตรวจสอบข้อมูลก่อนสร้างโมเดล**

ตรวจสอบว่า Feature ไม่มีค่า `NaN` หรือ `inf` และปรับสเกลด้วย `StandardScaler` ก่อนใช้ K-Means เพื่อให้ตัวแปรที่มีช่วงค่าต่างกันไม่ส่งผลต่อการคำนวณระยะห่างมากเกินไป

**ข้อควรระวัง:** Feature ที่ใช้ `end_x` และ `end_y` จะทราบค่าได้หลังเหตุการณ์สิ้นสุดแล้ว ดังนั้นชุด Feature นี้เหมาะกับการจัดกลุ่มรูปแบบการเคลื่อนที่ที่เกิดขึ้นแล้ว หากต้องการทำนายก่อนเหตุการณ์เกิดขึ้น ต้องตัด Feature ที่ต้องอาศัยจุดสิ้นสุดออก

```python
# ==========================================================
# 7. Feature Engineering
# ==========================================================
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GroupShuffleSplit

GOAL_X, GOAL_Y = 120.0, 40.0

# ใช้ข้อมูลหลัง Cleaning โดยไม่แก้ไข df_clean ต้นฉบับ
src = df_clean

sx = src["start_x"].to_numpy(dtype="float32")
sy = src["start_y"].to_numpy(dtype="float32")
ex = src["end_x"].to_numpy(dtype="float32")
ey = src["end_y"].to_numpy(dtype="float32")

# คำนวณการเปลี่ยนตำแหน่ง ระยะทาง และระยะห่างจากประตู
dx = ex - sx
dy = ey - sy
dist = np.hypot(dx, dy)

assert (dist > 0).all(), "พบระยะทาง 0 กรุณาตรวจสอบขั้นตอน Cleaning"

goal_start = np.hypot(GOAL_X - sx, GOAL_Y - sy)
goal_end = np.hypot(GOAL_X - ex, GOAL_Y - ey)

# ----------------------------------------------------------
# 1) สร้าง Features สำหรับ Clustering
# ----------------------------------------------------------
fe = pd.DataFrame({
    "start_x": sx,
    "start_y": sy,
    "log_dist": np.log1p(dist),
    "angle_sin": dy / dist,
    "angle_cos": dx / dist,
    "is_pass": (src["type"] == "Pass").to_numpy().astype("float32"),
    "goal_progress": goal_start - goal_end,
}).astype("float32")

FEATURES = list(fe.columns)

print("Features ที่ใช้:", FEATURES)
print(f"ขนาดข้อมูล: {fe.shape}")
print(f"หน่วยความจำ: {fe.memory_usage(deep=True).sum()/1e6:,.1f} MB")
display(fe.describe().T.round(3))

# ----------------------------------------------------------
# 2) เก็บตัวแปรสำหรับตีความผลภายหลัง
# ไม่ใช้เป็น Input ของ Clustering
# ----------------------------------------------------------
meta = src[["match_id", "team", "player", "type", "result"]].copy()
meta["dx"] = dx
meta["dy"] = dy
meta["dist"] = dist

# ----------------------------------------------------------
# 3) ตรวจสอบ Feature ที่ไม่ควรเข้าโมเดล
# ----------------------------------------------------------
FORBIDDEN = {"result", "team", "player", "match_id", "winner"}

assert FORBIDDEN.isdisjoint(FEATURES), \
    "พบตัวแปรผลการแข่งขันหรือตัวระบุหลุดเข้า Features"

# ตรวจสอบ NaN, inf และช่วงค่าของมุม
assert np.isfinite(fe.to_numpy()).all(), "พบ NaN หรือ inf"

assert fe["angle_sin"].between(-1, 1).all()
assert fe["angle_cos"].between(-1, 1).all()

assert len(fe) == len(src), "จำนวนแถวไม่ตรงกับข้อมูลต้นฉบับ"

print("ผ่านการตรวจสอบ Feature เบื้องต้น")

# ----------------------------------------------------------
# 4) ตรวจสอบว่า Feature แต่ละตัวรู้ค่าเมื่อใด
# ----------------------------------------------------------
timing = pd.DataFrame({
    "feature": FEATURES,
    "รู้ค่าเมื่อ": [
        "จุดเริ่ม",
        "จุดเริ่ม",
        "จุดจบ",
        "จุดจบ",
        "จุดจบ",
        "เมื่อทราบประเภท Event",
        "จุดจบ"
    ],
    "หมายเหตุ": [
        "พิกัดเริ่มต้นแกน X",
        "พิกัดเริ่มต้นแกน Y",
        "คำนวณจากพิกัดเริ่มต้นและปลายทาง",
        "คำนวณจากการเปลี่ยนแปลงตำแหน่ง",
        "คำนวณจากการเปลี่ยนแปลงตำแหน่ง",
        "Pass = 1, Carry = 0",
        "คำนวณระยะห่างจากจุดอ้างอิง (120, 40)"
    ]
})

display(timing)

# ----------------------------------------------------------
# 5) ตรวจสอบความสัมพันธ์ระหว่าง Features
# ใช้ตัวอย่าง 200,000 แถวเพื่อลดการใช้หน่วยความจำ
# ----------------------------------------------------------
corr = fe.sample(
    n=min(200_000, len(fe)),
    random_state=42
).corr().abs()

upper = corr.where(
    np.triu(np.ones(corr.shape, dtype=bool), k=1)
)

pairs = (
    upper.stack()
    .loc[lambda s: s > 0.7]
    .sort_values(ascending=False)
)

print("คู่ Feature ที่มี |correlation| > 0.7:")
print(pairs.round(3).to_string() if len(pairs) else "ไม่พบ")

# ----------------------------------------------------------
# 6) แบ่งข้อมูลด้วย GroupShuffleSplit เพื่อหลีกเลี่ยง Data Leakage ในระดับ Match
# ----------------------------------------------------------
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(
        fe[FEATURES],
        groups=meta["match_id"]
    )
)

# ----------------------------------------------------------
# 7) ปรับสเกล Features ก่อน Clustering
# ----------------------------------------------------------
# StandardScaler: fit เฉพาะชุด Train
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    fe.iloc[train_idx][FEATURES]
)

# ใช้ scaler เดิม transform ชุด Test
X_test_scaled = scaler.transform(
    fe.iloc[test_idx][FEATURES]
)
print("\nขนาดข้อมูล Train หลังปรับสเกล:", X_train_scaled.shape)
print("ค่าเฉลี่ย Train หลังปรับสเกล:", np.round(X_train_scaled.mean(axis=0), 3))
print("SD Train หลังปรับสเกล:", np.round(X_train_scaled.std(axis=0), 3))
```

`result`/`team` ใช้หลัง clustering เพื่อดู archetype mix ของทีมที่ชนะบ่อยเท่านั้น

## 8. Data Partitioning
อธิบายวิธีแบ่ง train / validation / test และเหตุผลที่วิธีนี้เหมาะกับข้อมูลและคำถามของกลุ่ม

**วิธีการแบ่งข้อมูล (Data Partitioning Method)**

* ใช้เทคนิค Group-based Split ผ่านโมดูล `GroupShuffleSplit` โดยกำหนดให้ `match_id` เป็นกลุ่ม (Group) ในการแบ่งชุดข้อมูล


* แบ่งข้อมูลออกเป็น Train 70%, Validation 15% และ Test 15% โดยสุ่มแยกที่ระดับ "แมตช์การแข่งขัน" แทนการสุ่มแยกทีละแถวเหตุการณ์ (Random Split)


* กระบวนการนี้รับประกันว่าเหตุการณ์การเคลื่อนที่ (Pass และ Carry) ทั้งหมดที่เกิดขึ้นในแมตช์เดียวกัน จะถูกจัดรวมไว้ในชุดข้อมูลใดชุดข้อมูลหนึ่งเพียงชุดเดียว ทำให้ไม่มีข้อมูลแมตช์ที่ซ้ำซ้อนกันระหว่างชุด Train, Validation และ Test



**เหตุผลที่วิธีนี้เหมาะกับข้อมูลและคำถาม**

* **การป้องกันข้อมูลรั่วไหล (Data Leakage):** ข้อมูลเหตุการณ์ฟุตบอลมีความสัมพันธ์และต่อเนื่องกันในแต่ละเกม การสุ่มแยกทีละแถวจะทำให้เหตุการณ์ที่ต่อเนื่องกันถูกแยกไปอยู่ทั้งชุด Train และ Test ส่งผลให้โมเดลจดจำรูปแบบเฉพาะของแมตช์นั้นและเกิด Overfitting การแบ่งด้วย `match_id` จึงช่วยป้องกันการรั่วไหลของข้อมูลภายในแมตช์ได้อย่างเด็ดขาด


* **การตอบโจทย์ด้านยุทธวิธี (Tactical Insights):** คำถามวิจัยของกลุ่มคือการศึกษา "รูปแบบการเคลื่อนที่ทำทาง (Movement Archetypes)" และสไตล์การเล่นของทีมที่มีอัตราการชนะสูง แท็กติกของทีมฟุตบอลจะเปลี่ยนไปตามแต่ละแมตช์ การทดสอบโมเดลด้วยแมตช์ที่โมเดลไม่เคยเห็นมาก่อนเลยทั้งเกม จะช่วยพิสูจน์ว่ารูปแบบการบุกที่เป็นแม่แบบสากลที่ค้นพบนั้นใช้งานได้จริงและสะท้อนโครงสร้างเกมบุกได้ตรงจุด


* **การประเมินความเสถียรของโมเดล Unsupervised Learning:** การจัดกลุ่มเพื่อสร้างตัวแปรเป้าหมาย (Target) หรือ Cluster ID ต้องการความเสถียร การแบ่งชุด Validation และ Test แยกออกมาเป็นแมตช์ใหม่ จะช่วยยืนยันว่าจุดศูนย์กลางของกลุ่ม (Centroids) ครอบคลุมรูปแบบการเคลื่อนที่ทั่วไป ไม่ได้เกิดจากการเรียนรู้จำเพาะกับชุดข้อมูล Train เพียงอย่างเดียว

```python
# ==========================================
# 8. Data Partitioning (Train 70% / Validation 15% / Test 15%)
# ==========================================
from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler

# 1. แบ่งรอบแรก: Train (70%) และ ชุดชั่วคราว Temp (30%)
splitter_train = GroupShuffleSplit(n_splits=1, test_size=0.30, random_state=42)
train_idx, temp_idx = next(
    splitter_train.split(fe[FEATURES], groups=meta["match_id"])
)

# 2. แบ่งรอบสอง: นำชุดชั่วคราว Temp (30%) มาแบ่งครึ่งเป็น Validation (15%) และ Test (15%)
temp_groups = meta.iloc[temp_idx]["match_id"]
splitter_val = GroupShuffleSplit(n_splits=1, test_size=0.50, random_state=42)

# ผลลัพธ์จะเป็น index สัมพัทธ์ของชุด temp
val_rel_idx, test_rel_idx = next(
    splitter_val.split(fe.iloc[temp_idx], groups=temp_groups)
)

# แปลงค่า index สัมพัทธ์กลับไปเป็น index ของ DataFrame ต้นฉบับ
val_idx = temp_idx[val_rel_idx]
test_idx = temp_idx[test_rel_idx]

# ==========================================
# รายงานผลการแบ่งข้อมูล
# ==========================================
print(f"จำนวนเหตุการณ์ใน Train: {len(train_idx):,} แถว (แมตช์ที่ไม่ซ้ำ: {meta.iloc[train_idx]['match_id'].nunique():,})")
print(f"จำนวนเหตุการณ์ใน Validation: {len(val_idx):,} แถว (แมตช์ที่ไม่ซ้ำ: {meta.iloc[val_idx]['match_id'].nunique():,})")
print(f"จำนวนเหตุการณ์ใน Test: {len(test_idx):,} แถว (แมตช์ที่ไม่ซ้ำ: {meta.iloc[test_idx]['match_id'].nunique():,})")

# ตรวจสอบ Data Leakage (ต้องเป็น 0 ทั้งหมด)
train_matches = set(meta.iloc[train_idx]["match_id"])
val_matches = set(meta.iloc[val_idx]["match_id"])
test_matches = set(meta.iloc[test_idx]["match_id"])

print("\nตรวจสอบ Data Leakage (จำนวนแมตช์ที่ซ้ำซ้อนกันข้ามชุดข้อมูล):")
print(f"Train vs Validation: {len(train_matches & val_matches)} แมตช์")
print(f"Train vs Test: {len(train_matches & test_matches)} แมตช์")
print(f"Validation vs Test: {len(val_matches & test_matches)} แมตช์")

# ==========================================
# 9. ปรับสเกลข้อมูล (Standardization) สำหรับเตรียมเข้าโมเดล
# ==========================================
scaler = StandardScaler()

# Fit และ Transform เฉพาะชุด Train เพื่อป้องกัน Data Leakage ด้านสถิติ
X_train_scaled = scaler.fit_transform(fe.iloc[train_idx][FEATURES])

# Transform ชุด Validation และ Test ด้วยค่าจาก Train
X_val_scaled = scaler.transform(fe.iloc[val_idx][FEATURES])
X_test_scaled = scaler.transform(fe.iloc[test_idx][FEATURES])

print("\nการปรับสเกลเสร็จสมบูรณ์ พร้อมนำ X_train_scaled เข้าโมเดล KMeans")
```

## 9. Preprocessing Pipeline
การจัดการ missing values, encoding, scaling ที่ต้อง "เรียนรู้จากข้อมูล" ให้ fit จาก **train set เท่านั้น**
แล้วนำไป transform ชุดอื่น (แนะนำใช้ `sklearn.pipeline.Pipeline` / `ColumnTransformer`)

```python
# ==========================================
# 9. Preprocessing Pipeline
# ==========================================
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
import pandas as pd

# 1. กำหนดกลุ่ม Features จากข้อมูลที่สร้างไว้ใน Feature Engineering
num_features = ['start_x', 'start_y', 'log_dist', 'angle_sin', 'angle_cos', 'goal_progress']
bin_features = ['is_pass']

# 2. สร้าง Pipeline สำหรับข้อมูลเชิงตัวเลข (Continuous Features)
# จัดการ Missing Values + ปรับสเกลข้อมูล (Standardization)
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),  # รองรับกรณีมีค่า NaN หลุดเข้ามา
    ('scaler', StandardScaler())                     # ปรับสเกลด้วย Mean=0, SD=1
])

# 3. สร้าง Pipeline สำหรับข้อมูล Binary
bin_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')) # กรณีมีค่า Missing ให้แทนด้วยค่าที่พบบ่อยสุด
])

# 4. รวม Pipeline ด้วย ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_pipeline, num_features),
        ('bin', bin_pipeline, bin_features)
    ],
    remainder='drop'
)

# ==========================================
# การ Fit จาก Train Set เท่านั้น แล้ว Transform ชุดอื่น
# ==========================================

# แยกชุดข้อมูล Feature ตาม Index จากขั้นตอน Data Partitioning
X_train = fe.iloc[train_idx][FEATURES]
X_val = fe.iloc[val_idx][FEATURES]
X_test = fe.iloc[test_idx][FEATURES]

# Fit และ Transform เฉพาะ Train Set
X_train_preprocessed = preprocessor.fit_transform(X_train)

# Transform ชุด Validation และ Test โดยใช้ค่าสถิติ (Mean, SD) ที่เรียนรู้จาก Train Set เท่านั้น
X_val_preprocessed = preprocessor.transform(X_val)
X_test_preprocessed = preprocessor.transform(X_test)

# รายงานผลลัพธ์
print("Preprocessing Pipeline ดำเนินการสำเร็จ!")
print(f"ขนาดข้อมูล X_train_preprocessed: {X_train_preprocessed.shape}")
print(f"ขนาดข้อมูล X_val_preprocessed:   {X_val_preprocessed.shape}")
print(f"ขนาดข้อมูล X_test_preprocessed:  {X_test_preprocessed.shape}")
```
